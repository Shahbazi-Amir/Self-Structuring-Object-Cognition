## سند شروع پروژه: SSOC — Self-Structuring Object Cognition

### 1) این بحث از کجا شروع شد و چرا به SSOC رسید؟

**شروع بحث:**
بحث با سؤالِ «دانش فطری در انسان چیست و چرا AI شبیه انسان استدلال نمی‌کند؟» شروع شد. در مسیر گفتگو مشخص شد که مدل‌های عصبی/LLMهای امروز (مثل GPTها) عمدتاً **پردازش‌گرهای آماریِ زبان** هستند:

* در تولید متن قوی‌اند
* اما در **پایبندی پایدار به اصول منطقی/بدیهیات** و «فهمِ ساختاری» ضعف دارند
* تناقض را ممکن است “ادامه بدهند” و بدون توقف، پاسخ بسازند

**وسط راه چه شد؟**
برای ریشه‌یابی، روی خط چامسکی (UG/LAD، Poverty of the Stimulus) رفتیم:

* کودک با داده کم یاد می‌گیرد → احتمال وجود یک «هسته/مکانیزم ذاتی»
* نتیجه عملی برای AI: شاید به جای افزایش داده و پارامتر، باید یک **هسته‌ی شناختی/منطقی** داشته باشیم که رفتار مدل را محدود و هدایت کند.

همزمان چند خانواده ایده بررسی شد:

* معماری‌های شناختی (مثل Global Workspace/LIDA)
* خودسازمان‌دهی مفهومی (Ouroboros/ReSOM)
* گراف مفهومی + قوانین
* لایه بازتاب (Reflective Layer)
  و مشکل مشترک این بود:
* یا **خیلی فلسفی و غیرقابل تست** می‌شوند
* یا **خیلی دستی و غیرمقیاس‌پذیر**
* یا **زبان طبیعی روان ندارند**

**چرا به “پارادایم جدید” رسیدیم؟**
چون هدف فقط «بهبود یک چت‌بات» نبود؛ هدف این بود که سیستم بتواند:

* با داده کم رشد کند
* “ساختار مفهومی” بسازد
* اصول بدیهی را حفظ کند
* خودش خودش را اصلاح کند
  این‌ها با پارادایم غالبِ «مدل بزرگ آماری» به‌تنهایی تضمین‌پذیر نیستند.
  بنابراین تصمیم شد یک پارادایم جدید تعریف شود که پایه‌اش **Object-based cognition** باشد:
  یعنی شناخت از “شیء/شیء‌واره” و روابط و قیود شروع شود، نه از توکن و هم‌وقوعی آماری.

---

### 2) SSOC دقیقاً چیست؟ (تعریف کامل و عملی)

**SSOC = Self-Structuring Object Cognition**

> یک معماری شناختی که دانش را به صورت «اشیاء پایدار» (Object) با **هویت (Identity)**، **ویژگی‌ها (Properties)**، **روابط (Relations)** و **قیود (Constraints)** نمایش می‌دهد و می‌تواند ساختار داخلی خودش را با مکانیزم‌های بازتابی **خودشکل‌دهی/بازنویسی** کند.

#### 2.1) فرق SSOC با LLMها در یک جمله

* LLM: *Token/Pattern-based text generator*
* SSOC: *Object/Constraint-based cognition engine (با اتصال زبان به عنوان رابط)*

#### 2.2) چرا “Object” محور است؟

چون در شناخت انسانی، واحدهای پایدار مانند «این لیوان»، «آن آدم»، «قانون»، «هدف»، «علت» نقش بنیادی دارند. زبان روی این واحدها سوار می‌شود.
در SSOC، زبان فقط یک لایه رابط است؛ هسته شناختی باید بدون زبان هم معنا داشته باشد.

---

### 3) این کار چه به درد می‌خورد؟ (ارزش/کاربرد)

اگر SSOC موفق شود، خروجی‌های بالقوه‌اش:

1. **کاهش تناقض و هذیان در سیستم‌های زبانی**
   به جای امید به “خوش‌شانس بودن” خروجی، سیستم با قیود صوری جلوی تناقض را می‌گیرد.

2. **قابلیت “فهم ساختاری” و توضیح‌پذیری**
   چون دانش به شکل گراف اشیاء و قیود است، می‌شود نشان داد «چرا» نتیجه این شد.

3. **یادگیری با داده کمتر (کم‌نمونه) از طریق ساختار**
   به جای جذب الگو از میلیون‌ها نمونه، بخشی از یادگیری از طریق ساختن مدل مفهومی و قیود انجام می‌شود.

4. **پایه برای پژوهش دانشگاهی / مسیر PhD**
   چون هم formalization دارد، هم prototype، هم benchmark، هم امکان مقاله‌پذیری.

---

## 🔴 مشکل دوم: انفجار پیچیدگی (Complexity Explosion)

وقتی بگویی:

> قیود + بازنویسی + متا-قیود + ساخت مفهوم + خودساختاری

سیستم وارد قلمرو NP-hard و حتی بدتر می‌شود.

مثلاً:

* Consistency checking در logic کامل → گاهی undecidable
* Abductive repair → combinatorial explosion
* Identity resolution → نزدیک به entity matching سنگین

اگر این‌ها را کامل بخواهی، پروژه از کنترل خارج می‌شود.

---

## 🔴 مشکل سوم: Self-Structuring واقعی خیلی سخت‌تر از چیزی است که فکر می‌کنی

ساخت hierarchy از داده؟
Concept induction؟
Meta-constraints؟

این‌ها نزدیک‌اند به:

* Inductive Logic Programming
* Program synthesis
* Structure learning in probabilistic graphs

که هرکدام حوزه دکتری کامل‌اند.

---

## 🔴 مشکل چهارم: زبان

LLM را اگر فقط parser کنی،
تو وابسته‌ای به مدلی که خودش inconsistent است.

اگر parser deterministic بسازی،
coverage زبان خیلی محدود می‌شود.

این گلوگاه جدی است.

---

## ✅ SSOC Minimal Viable Architecture

### لایه 1 — Structured Memory (قابل ساخت در 6–9 ماه)

* Typed objects
* Relations
* Hard constraints فقط نوع ساده:

  * mutual exclusion
  * type violation
  * direct negation
* Graph storage

هیچ self-structuring در این مرحله.

---

### لایه 2 — Consistency Gate

* هر خروجی LLM → parse ساده
* تبدیل به triples محدود
* بررسی تناقض با حافظه
* اگر تناقض:

  * توقف
  * درخواست clarification
  * یا پیشنهاد اصلاح

این قابل اجراست.

---

### لایه 3 — Limited Rewrite (نه بازنویسی عمومی)

به جای:

> rewrite آزاد

فقط این‌ها را بده:

* merge identical objects
* create subtype from repeated pattern
* compress repeated relations

نه meta-constraint.
نه induction عمومی.
