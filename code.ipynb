````python
import networkx as nx
from pprint import pprint

# ----------------------------
# لایه ادراک اولیه (Perception Layer)
# ----------------------------
class PerceptionLayer:
    def __init__(self):
        self.raw_data = []

    def feed(self, data):
        """داده خام را وارد می‌کند"""
        self.raw_data = data
        print("[Perception] داده ورودی دریافت شد:", self.raw_data)

# ----------------------------
# لایه ساخت ابژه (Object Construct Layer)
# ----------------------------
class ObjectConstructLayer:
    def __init__(self):
        self.objects = {}

    def create_objects(self, raw_data):
        """هر عنصر داده را به یک ابژه تبدیل می‌کند"""
        for item in raw_data:
            self.objects[item] = {"type": "object", "relations": {}}
        print("[Object Construct] ابژه‌ها ساخته شدند:")
        pprint(self.objects)
        return self.objects

# ----------------------------
# موتور معنایی (Semantic Engine)
# ----------------------------
class SemanticEngine:
    def __init__(self):
        self.graph = nx.DiGraph()

    def build_relations(self, objects):
        """روابط بین ابژه‌ها را تشخیص می‌دهد"""
        items = list(objects.keys())
        for i in range(len(items)):
            for j in range(i + 1, len(items)):
                self.graph.add_edge(items[i], items[j], relation="part_of" if i < j else "contains")
        print("[Semantic Engine] روابط بین ابژه‌ها ساخته شد.")
        return self.graph

    def visualize_graph(self):
        """نمایش گراف معنایی"""
        print("\n[Semantic Engine] گراف معنایی:")
        for u, v, d in self.graph.edges(data=True):
            print(f"{u} → {v} | Relation: {d['relation']}")

# ----------------------------
# هسته بازتابی (Reflective Core)
# ----------------------------
class ReflectiveCore:
    def __init__(self, graph):
        self.graph = graph

    def reflect(self):
        """تحلیل ساختاری گراف و به‌روزرسانی آن"""
        print("[Reflective Core] بازتاب انجام شد.")
        # مثال ساده: تشخیص گره‌هایی که رابطه دارند
        degrees = dict(self.graph.degree)
        most_connected = max(degrees, key=degrees.get)
        print(f"[Reflective Core] مفهوم مرکزی: '{most_connected}' (با بیشترین ارتباط)")

        # افزودن یک مفهوم منطقی ساده: "کل > جز"
        if "Whole" in self.graph.nodes and "Part" in self.graph.nodes:
            print("[Reflective Core] مفهوم منطقی: 'کل > جز' تشخیص داده شد.")

# ----------------------------
# لایه کنشی (Action Layer)
# ----------------------------
class ActionLayer:
    def __init__(self, graph):
        self.graph = graph

    def query(self, source, target):
        """بررسی وجود رابطه بین دو مفهوم"""
        if self.graph.has_edge(source, target):
            rel = self.graph[source][target]['relation']
            print(f"[Action Layer] رابطه بین '{source}' و '{target}': {rel}")
        else:
            print(f"[Action Layer] رابطه‌ای بین '{source}' و '{target}' یافت نشد.")

# ----------------------------
# معماری کامل SSOC
# ----------------------------
def run_ssoc_pipeline(input_data):
    print("🚀 شروع فرآیند Self-Structuring Object Cognition\n")

    perception = PerceptionLayer()
    perception.feed(input_data)

    obj_construct = ObjectConstructLayer()
    objects = obj_construct.create_objects(perception.raw_data)

    sem_engine = SemanticEngine()
    semantic_graph = sem_engine.build_relations(objects)
    sem_engine.visualize_graph()

    reflective_core = ReflectiveCore(semantic_graph)
    reflective_core.reflect()

    action_layer = ActionLayer(semantic_graph)
    action_layer.query("A", "B")
    action_layer.query("C", "A")

    print("\n✅ فرآیند SSOC به پایان رسید.")

# ----------------------------
# مثال ورودی
# ----------------------------
input_data = ["A", "B", "C"]
run_ssoc_pipeline(input_data)
````

### ✅ سؤال آخر: آیا این کد یک مدل واقعی از SSOC است؟

بله، این کد یک **نسخه مفهومی و ساده‌شده از Self-Structuring Object Cognition** است.  
اما هنوز:
- **قدرت استنتاج عمیق** ندارد،
- **خودبازنویسی قواعد** وجود ندارد،
- **تعامل با داده واقعی** نیست.

````python
class SSOC_Minimal:
    def __init__(self):
        self.objects = []
        self.relations = []

    def perceive(self, raw_input):
        print("👁️ داده ورودی:", raw_input)
        self.objects = raw_input

    def structure(self):
        for i in range(len(self.objects) - 1):
            relation = (self.objects[i], "next_to", self.objects[i + 1])
            self.relations.append(relation)
        print("🔗 روابط ساخته شده:")
        for r in self.relations:
            print(f"{r[0]} --[{r[1]}]--> {r[2]}")

    def reflect(self):
        print("🔍 بازتاب: تعداد مفاهیم =", len(self.objects))
        print("🔍 بازتاب: رابطه‌ها =", len(self.relations))

    def act(self):
        print("🤖 کنش: بررسی اگر A با B مرتبطه")
        for r in self.relations:
            if r[0] == "A" and r[2] == "B":
                print("✅ A با B مرتبط است.")
                return
        print("❌ رابطه‌ای بین A و B نیست.")

ssoc = SSOC_Minimal()
ssoc.perceive(["A", "B", "C"])
ssoc.structure()
ssoc.reflect()
ssoc.act()
````
