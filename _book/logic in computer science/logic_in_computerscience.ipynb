### **فصل اول: مقدمه‌ای بر منطق گزاره‌ای (Propositional Logic)**  
کتاب **"منطق در علوم کامپیوتر" (Huth & Ryan)** در فصل اول به معرفی مفاهیم پایه‌ای **منطق گزاره‌ای** (Propositional Logic) می‌پردازد. این فصل، بنیادی‌ترین بخش کتاب است و به شما می‌آموزد که چگونه از منطق برای نمایش و استدلال درباره اطلاعات استفاده کنید. در ادامه، توضیح کاملی از این فصل ارائه می‌شود.

---

## **1. مقدمه**
- **هدف**: این فصل به شما می‌آموزد که چگونه با استفاده از **گزاره‌ها** و **رابط‌های منطقی**، اطلاعات را به صورت ریاضی مدل کنید و آنها را تحلیل کنید.
- **کاربردها**: منطق گزاره‌ای در علوم کامپیوتر برای طراحی مدارهای دیجیتال، برنامه‌نویسی، و اثبات الگوریتم‌ها استفاده می‌شود.

---

## **2. گزاره‌ها (Propositions)**
- **تعریف**:  
  یک **گزاره** جمله‌ای است که می‌تواند **درست** (True) یا **نادرست** (False) باشد، اما هر دو را همزمان ندارد.  
  - **مثال‌های گزاره**:
    - "زمین دور خورشید می‌چرخد." → **درست**
    - "عدد ۲ زوج است." → **درست**
    - "۳ بزرگ‌تر از ۵ است." → **نادرست**
  - **مثال‌های غیرگزاره**:
    - "سلام!" → این جمله نه درست است و نه نادرست.
    - "x بزرگ‌تر از ۱۰ است." → این جمله بدون مشخص بودن مقدار `x` قابل ارزیابی نیست.

---

## **3. رابط‌های منطقی (Logical Connectives)**
رابط‌های منطقی ابزارهایی هستند که از آنها برای ترکیب گزاره‌ها و ساختن فرمول‌های پیچیده‌تر استفاده می‌شود. مهم‌ترین رابط‌های منطقی عبارتند از:

### **الف) نقیض (Negation): `¬`**
- **تعریف**: نقیض یک گزاره، مخالف آن گزاره است.  
  - اگر `P` یک گزاره باشد، `¬P` بیان می‌کند که `P` نادرست است.
- **جدول درستی**:
  | P   | ¬P  |
  |-----|-----|
  | T   | F   |
  | F   | T   |

### **ب) عطف (Conjunction): `∧`**
- **تعریف**: عطف دو گزاره (`P ∧ Q`) فقط وقتی درست است که هر دو گزاره درست باشند.
- **جدول درستی**:
  | P   | Q   | P ∧ Q |
  |-----|-----|-------|
  | T   | T   | T     |
  | T   | F   | F     |
  | F   | T   | F     |
  | F   | F   | F     |

### **ج) فصل (Disjunction): `∨`**
- **تعریف**: فصل دو گزاره (`P ∨ Q`) فقط وقتی نادرست است که هر دو گزاره نادرست باشند.
- **جدول درستی**:
  | P   | Q   | P ∨ Q |
  |-----|-----|-------|
  | T   | T   | T     |
  | T   | F   | T     |
  | F   | T   | T     |
  | F   | F   | F     |

### **د) شرطی (Implication): `→`**
- **تعریف**: شرطی (`P → Q`) بیان می‌کند که اگر `P` درست باشد، آنگاه `Q` نیز درست است.  
  - اگر `P` نادرست باشد، `P → Q` همواره درست است.
- **جدول درستی**:
  | P   | Q   | P → Q |
  |-----|-----|-------|
  | T   | T   | T     |
  | T   | F   | F     |
  | F   | T   | T     |
  | F   | F   | T     |

### **هـ) دوشرطی (Biconditional): `↔`**
- **تعریف**: دوشرطی (`P ↔ Q`) بیان می‌کند که `P` و `Q` هم‌ارز هستند، یعنی هر دو درست یا هر دو نادرست باشند.
- **جدول درستی**:
  | P   | Q   | P ↔ Q |
  |-----|-----|-------|
  | T   | T   | T     |
  | T   | F   | F     |
  | F   | T   | F     |
  | F   | F   | T     |

---

## **4. جدول درستی (Truth Table)**
- **تعریف**: جدول درستی ابزاری است که برای محاسبه ارزش درستی یک فرمول منطقی بر اساس تمام ترکیب‌های ممکن از ارزش‌های گزاره‌ها استفاده می‌شود.
- **مثال**:  
  فرض کنید فرمول منطقی `P ∧ (Q ∨ ¬R)` داریم. جدول درستی آن به صورت زیر است:
  | P   | Q   | R   | ¬R  | Q ∨ ¬R | P ∧ (Q ∨ ¬R) |
  |-----|-----|-----|-----|--------|--------------|
  | T   | T   | T   | F   | T      | T            |
  | T   | T   | F   | T   | T      | T            |
  | T   | F   | T   | F   | F      | F            |
  | T   | F   | F   | T   | T      | T            |
  | F   | T   | T   | F   | T      | F            |
  | F   | T   | F   | T   | T      | F            |
  | F   | F   | T   | F   | F      | F            |
  | F   | F   | F   | T   | T      | F            |

---

## **5. معادل‌سازی منطقی (Logical Equivalence)**
- **تعریف**: دو فرمول منطقی معادل هستند اگر در تمام حالات ارزش درستی یکسانی داشته باشند.  
  - نماد: `≡`
- **قوانین معروف معادل‌سازی**:
  - **قانون دمورگان (De Morgan's Laws)**:
    - `¬(P ∧ Q) ≡ ¬P ∨ ¬Q`
    - `¬(P ∨ Q) ≡ ¬P ∧ ¬Q`
  - **توزیع‌پذیری (Distributivity)**:
    - `P ∧ (Q ∨ R) ≡ (P ∧ Q) ∨ (P ∧ R)`
    - `P ∨ (Q ∧ R) ≡ (P ∨ Q) ∧ (P ∨ R)`
  - **نقیض دوگانه (Double Negation)**:
    - `¬(¬P) ≡ P`

---

## **6. استنتاج منطقی (Logical Reasoning)**
- **تعریف**: استنتاج منطقی فرآیندی است که از مجموعه‌ای از فرضیات (Premises) به یک نتیجه (Conclusion) می‌رسد.
- **مثال‌های قواعد استنتاج**:
  - **Modus Ponens**:  
    اگر `P → Q` و `P` درست باشد، آنگاه `Q` نیز درست است.
  - **Modus Tollens**:  
    اگر `P → Q` و `¬Q` درست باشد، آنگاه `¬P` نیز درست است.

---

## **7. کاربردها در علوم کامپیوتر**
- **مدارهای دیجیتال**: منطق گزاره‌ای پایه‌ای برای طراحی مدارهای منطقی (Logic Gates) است.
- **برنامه‌نویسی**: شرط‌ها و حلقه‌ها در برنامه‌نویسی از منطق گزاره‌ای استفاده می‌کنند.
- **اثبات‌های ریاضی**: منطق گزاره‌ای برای اثبات قضایای ریاضی و الگوریتم‌ها استفاده می‌شود.

---

### **نتیجه‌گیری**
فصل اول کتاب Huth & Ryan به شما می‌آموزد که چگونه از گزاره‌ها و رابط‌های منطقی برای مدل‌سازی اطلاعات استفاده کنید. این مفاهیم پایه‌ای برای درک منطق محمولات (Predicate Logic) و سایر بخش‌های کتاب هستند.
