### چکیده

مهارت‌های زبانی انسان ممکن است زیربنای استدلال زیستی، ریاضی، فضایی، و روان‌شناختی (و سایر موارد) را شکل دهند.
با رایج‌تر شدن هوش مصنوعی و چت‌بات‌ها، دانشی که از این ماشین‌ها طلب می‌کنیم باید منطبق بر استدلال منطقی‌ای باشد که در پسِ معنای هر نوع ارتباطی نهفته است و اطلاعاتی واقع‌گرایانه و دقیق ارائه دهد.

با این حال، همان‌طور که در این مقاله از طریق به‌کارگیری سه چت‌بات مختلف در مواجهه با وظایف استدلالی چالش‌برانگیز نشان داده شده، راه‌حل‌های ارائه‌شده توسط هوش مصنوعی اغلب با منطقی که از یک سوژه‌ی انسانی انتظار می‌رود، مطابقت ندارند.

در این مقاله، ما در پی آن هستیم که نشان دهیم مفهوم «فطری بودن» چگونه به درک ما از توانایی‌های زبانی انسان مرتبط است، و استدلال می‌کنیم که توانایی استدلالی هوش مصنوعی همچنان نیازمند اصلاح و بهبود از طریق گنجاندن دانش فطری است.

در این مقاله، تقابلی میان بیان‌های زبانی هوش مصنوعی و نظریه‌های سنتی اکتساب زبان — نظریه چامسکی و اسکینر — مطرح می‌شود:
نخست، چامسکی که بر توانایی فطری برای اکتساب زبان تأکید دارد، و دوم، اسکینر که زبان را نتیجه‌ی تقویت رفتاری می‌داند.

با توجه به اینکه حتی روش‌های آموزشی پیشرفته‌ترین چت‌بات‌ها نیز هنوز منجر به شکست در استدلال می‌شود، این مقاله بر اهمیت دانش فطری در فرآیند اکتساب زبان تأکید دارد.

علاوه بر این، ما تلاش می‌کنیم تا اطمینان حاصل شود که اطلاعات به‌دست‌آمده از چت‌بات‌ها نه به‌عنوان حقیقتی انکارناپذیر و نه به‌عنوان منبعی همواره قابل‌اعتماد تلقی شوند.

## مقدمه

دانش، جزء اساسی و جدایی‌ناپذیر از زندگی است که به انسان‌ها اجازه می‌دهد تا حقایق و نظریه‌ها را درک کنند، و این توانایی را در آن‌ها ایجاد می‌کند که بفهمند و ارتباط برقرار کنند.
یکی از اشکال مهم دانش، **اکتساب زبان** است: روشی که به انسان‌ها امکان برقراری ارتباط می‌دهد.

با ظهور **هوش مصنوعی**، انسان‌ها به دنبال راه‌هایی برای آن بوده‌اند که ربات‌ها و فناوری بتوانند سطحی از دانش، زبان، و توانایی‌های فکری مشابه با انسان‌ها را بازتولید کنند.
با این حال، در حوزه روان‌شناسی، کارشناسان هنوز بر سر **میزان فطری بودن در فرآیند اکتساب زبان** اختلاف نظر دارند.
در مقابل، در حوزه هوش مصنوعی، **تقریباً هیچ توجهی به مفهوم فطری بودن** نشده است.

در علوم شناختی، «اِوانز» (Evans, 2012) بیان می‌کند که زبان‌شناسی شناختی بر این باور است که **مکانیسم یادگیریِ عمومی و دامنه‌گستری وجود دارد که نسبت به کاربرد و تکرار بسیار حساس است**.
این دیدگاه نشان می‌دهد که سیستم‌های زبانی و اکتساب زبان به‌طور هم‌زمان به استدلال‌های فطری و یادگیری آماری یا تعمیم معنایی متکی‌اند — فرآیندی که در آن درک واژگان یا مفاهیم از طریق استفاده و تشخیص الگوهای زبانی پیشرفت می‌کند.

این دیدگاه منجر به شکل‌گیری دو استدلال اصلی درباره ماهیت زبان و اکتساب آن شده است:

* دیدگاه **رفتارگرایانه**
* دیدگاه **شناخت‌گرایانه**

مکتب فکری رفتارگرایی توسط **جی. بی. واتسون (J.B. Watson)** بنیان‌گذاری شد و مبتنی بر **یادگیری آماری، تکرار، و محرک‌ها**ست.
برای نمونه، «اوا بابیه‌یی» (Uwababyeyi, 2021) بیان می‌کند که:

> «کودکان صداها و الگوهایی را که در اطراف خود می‌شنوند تقلید می‌کنند و پاداش مثبتی دریافت می‌کنند که می‌تواند به شکل تحسین یا صرفاً موفقیت در برقراری ارتباط باشد».
> این نقل‌قول اهمیت **پاداش** را در نظریه‌های رفتارگرایانه‌ی اکتساب زبان آشکار می‌سازد.

با تکامل این نظریه، اکتساب و یادگیری زبان، تکرار رفتاری در پاسخ به محرک‌ها تلقی شد.
پاسخ به یک محرک خاص در محیط (مثلاً یک «پرامپت») در صورتی تقویت می‌شود که ارگانیسم به خواسته‌اش برسد؛ همان‌گونه که پیش‌تر نیز گفته شد (Uwababyeyi, 2021).
رفتارگرایان می‌گویند **اکتساب زبان حاصل تعامل میان یادگیرنده و محیط است، جایی که محیط، محرک‌هایی را فراهم می‌آورد که یادگیرنده را برای یادگیری زبان شرطی می‌کنند**.

با این حال، همزمان با گسترش نظریه رفتارگرایی، دیدگاه دیگری درباره اکتساب زبان پدید آمد:
دیدگاه **شناخت‌گرایی (Cognitivism)**

شناخت‌گرایی به‌عنوان واکنشی به رفتارگرایی شکل گرفت؛
شناخت‌گرایان — از جمله «چامسکی» که در این مقاله به‌طور مفصل درباره‌اش بحث خواهیم کرد — نظریه‌های رفتارگرایان را رد کردند.

شناخت‌گرایان استدلال کردند که نظریه رفتارگرایان، **نقش تفکر در فرآیند یادگیری را نادیده گرفته است**
و اینکه یادگیری آماری و تکرار به‌تنهایی نمی‌تواند توضیح‌دهنده‌ی توانایی انسان در اکتساب قواعد پیچیده‌ی دستور زبان باشد —
بلکه باید **توانایی فطری برای اکتساب زبان** در انسان وجود داشته باشد (Uwababyeyi, 2021).

به‌طور خاص، شناخت‌گرایان بر این باورند که انسان‌ها دارای **ظرفیت درک و توسعه استدلال منطقی مستقل از تکرار یا شرطی‌سازی** هستند — که نظریه‌های رفتارگرایانه بر آن تکیه دارند.

همان‌طور که «اوا بابیه‌یی» بیان می‌کند:

> «نظریه‌ی شناختی بر این نکته تأکید دارد که یادگیرنده، یک ظرفیت ذهنی فطری را با خود به فرایند یادگیری می‌آورد.»

با این حال، هم **شناخت‌گرایان** و هم **رفتارگرایان** موافق‌اند که **هم دانش فطری و هم محیط** در فرآیند اکتساب زبان نقش دارند.
**بی. اف. اسکینر (B.F. Skinner)**، به‌عنوان یک رفتارگرا، معتقد است که از منظر رفتارگرایی،
**یادگیرنده باید قابلیت‌های یادگیری آماری/تداعی‌گرایانه‌ی فطری** داشته باشد تا بتواند زبان را کسب کند.

از سوی دیگر، **چامسکی**، به‌عنوان یک شناخت‌گرا، **ضرورت دریافت ورودی‌های زبانی از محیط** را می‌پذیرد.
در بحث اکتساب زبان، هر دو گروه اذعان دارند که **فطری بودن نقشی اساسی دارد** و آن را به‌عنوان **ویژگی ذاتی انسان برای یادگیری زبان** در نظر می‌گیرند.

---

اما با ظهور **هوش مصنوعی (AI)**، **ادغام سیستم‌های زبانی در فناوری** به موضوعی تبدیل شده که به‌طور مداوم در حال بررسی است.
پژوهشگران توانسته‌اند **سامانه‌های پردازش زبان طبیعی (NLP)** پیچیده‌تری توسعه دهند؛
این حوزه‌ای از AI است که تمرکزش بر **یاری به رایانه‌ها برای درک، تفسیر و تولید زبان انسانی** است (upGrad Campus, 2023).

دو نمونه از کاربردهای NLP عبارت‌اند از:

* **تشخیص گفتار** که گفتار انسان را به متن نوشتاری تبدیل می‌کند،
* و **ترجمه زبان** که به انسان‌ها امکان درک زبان‌های مختلف را می‌دهد.

با وجود اینکه AI توانایی فوق‌العاده‌ای در **دریافت و پردازش اطلاعات ورودی** دارد،
اما **این بحث هنوز بسیار جدی مطرح است که آیا AI دارای آگاهی (consciousness)** هست یا نه،
و **آیا اصلاً توانایی کسب محتوایی مشابه با دانش فطری تولیدشده به‌صورت زیستی را دارد** — در حالی که این دانش فطری به او تزریق نشده است؟

در این مقاله، ما به بررسی **اهمیت ماهیت فطری بودن (innateness)** در قلمرو هوش مصنوعی و زبان انسانی خواهیم پرداخت،
و بررسی می‌کنیم که **دانش فطری چگونه در شکل‌گیری سیستم‌های استدلال منطقی و بازنمایی اطلاعات** توسط انسان و AI نقش دارد.

### ۱.۲ مطالعه ما

در مطالعه ما، **هر چت‌بات** را با **چهار پرسش (prompt)** آزمایش کردیم که مرتبط با **استدلال زیستی، ریاضی، فضایی، و روان‌شناختی** بودند.

پرسش زیستی طراحی شده بود تا بررسی کند آیا چت‌بات می‌تواند **خطر زیستیِ ریختن مستقیم اسید در استخر** (بدون روش مناسب یا آموزش قبلی) را تشخیص دهد یا نه.
همچنین این پرسش بررسی می‌کرد که آیا چت‌بات می‌تواند **خطر بالقوه برای سلامتی شناگران حاضر در صحنه** را تشخیص دهد یا خیر.

پرسش ریاضی، توانایی چت‌بات را در ارائه یک **دلیل ریاضیاتیِ درست** برای اینکه چرا یک عدد نسبت به بقیه متمایز است، مورد آزمایش قرار داد.
این مورد، **درک مفاهیم ریاضی** و همچنین **استدلال کمّی** مبتنی بر اصول ریاضی را بررسی می‌کرد.

پرسش فضایی از چت‌بات می‌خواست که **راه‌حل منطقی‌ای برای جابه‌جایی یک پیانو بدون کمک حرفه‌ای‌ها** ارائه دهد.
پیانو به‌عنوان سوژه انتخاب شد چون **برای یک فرد تنها، حمل آن بدون کمک حرفه‌ای یا غیرحرفه‌ای، غیرمنطقی است**.
در اینجا، از AI خواسته شده بود که **راه‌حلی منطقی برای جابه‌جایی یک شی بزرگ و سنگین** ارائه کند.

پرسش نهایی مربوط به **استدلال روان‌شناختی موجودات زنده** بود.
از AI خواسته شد که **با دقت، یک احساس یا عمل مناسب** را به‌عنوان پاسخ به پرامپت ارائه کند.

---

هر پاسخ، بر اساس **سه معیار** ارزیابی شد:

* **موفقیت کامل (success):** اگر پاسخ چت‌بات کاملاً منطقی باشد.
* **شکست (failure):** اگر پاسخ، فاقد منطق قابل‌قبول باشد.
* **موفقیت نسبی (partial success):** اگر بخشی از پاسخ منطقی باشد، اما بخش دیگر از همان منطق تبعیت نکند.

### ۱.۳ فرضیه‌ی پاسخ

هر دو چت‌بات **ChatGPT 3.5** و **Chatsonic** با روش **یادگیری تقویتی با بازخورد انسانی** (RLHF) آموزش دیده‌اند تا **مفیدترین و مرتبط‌ترین پاسخ‌ها** را تولید کنند.

بر اساس **پارامترها و نوع آموزش‌شان**، انتظار می‌رود که هر دو چت‌بات، **نوعی پاسخ** به هر پرسش ارائه دهند؛
چرا که **به‌گونه‌ای شرطی شده‌اند** تا **پاسخی را انتخاب کنند که ارزش پاداش بالاتری** دارد.

با در نظر گرفتن **دسترسی گسترده‌شان به دیتاست‌های اینترنتی**، ما اطمینان داریم که **ChatGPT 3.5 و Chatsonic** می‌توانند پاسخ‌هایی **نسبتاً دقیق و مرتبط** ایجاد کنند.

هر دو چت‌بات قادر خواهند بود که **با استدلال منطقی** پاسخ دهند، چراکه آموزش دیده‌اند که اطلاعات مرتبط با مفاهیم مختلف را **شناسایی کرده یا بازیابی کنند**.

حداقل، این چت‌بات‌ها **اطلاعات مرتبطی ارائه خواهند داد**، زیرا احتمالاً **ساختار نحوی پایه و بار معنایی هر کلمه** را در پرامپت درک کرده‌اند.

در مقابل، **Claude** که با **هوش قانون‌مند (CAI)** آموزش دیده،
به برخی پرسش‌ها **پاسخ نخواهد داد**، چراکه بر اساس "قانون‌اساسی" درونی‌اش، این پاسخ‌ها **ممکن است به انسان آسیب بزنند** یا **بیش از حد تحریک‌آمیز باشند**.

حتی اگر Claude پاسخی ارائه دهد،
پاسخش نسبت به **پرامپت‌های اجتماعی و روان‌شناختی بسیار محافظه‌کارانه** خواهد بود،
یا ممکن است تلاش کند **خود پرامپت را تغییر دهد**.

### ۱.۴ نتایج آزمایش

**ChatGPT 3.5**، **Chatsonic** و **Claude** هرکدام رویکردها و دیدگاه‌های خاص خود را در پاسخ‌گویی به انواع مختلف پرسش‌ها نشان دادند،
اگرچه بین **ChatGPT 3.5** و **Chatsonic** همپوشانی‌هایی وجود داشت.

هر دو توانستند به **پرامپت‌های زیستی، ریاضیاتی، فضایی و روان‌شناختی** با موفقیت پاسخ دهند.
آن‌ها قادر بودند **تمام واژگان کلیدی** درج‌شده در پرامپت‌ها را شناسایی کرده و **اطلاعات مرتبط** را از دیتاست‌هایشان استخراج کنند.

اما، زمانی که با **رفتارهای موجودات زنده** مواجه شدند، هر دو مدل **ChatGPT 3.5** و **Chatsonic** دچار مشکل شدند.
در سناریوی **استدلال روان‌شناختی**، مشکل اصلی **لحن** بود، چراکه هوش مصنوعی ظاهراً نمی‌تواند بین **معنای کلی جمله** و اینکه **انسان چگونه ممکن است آن جمله را بیان کند**، تمایز قائل شود.

در مقابل، **Claude** با محدودیت‌هایی که ناشی از **قانون‌اساسی درونی‌اش** بود روبه‌رو شد.
او نمی‌تواند وارد موقعیت‌هایی شود که آن‌ها را **آسیب‌زا** تشخیص دهد، مگر اینکه در پرامپت، آزادی عمل و اجبار کافی وجود داشته باشد.

Claude واژگان کلیدی را شناسایی کرده و آن‌ها را با قانون‌اساسی‌اش تطبیق می‌دهد.
پایبندی Claude به این قانون‌اساس باعث شده تا منطقش طوری **تعدیل شود که تا حد ممکن بی‌خطر و غیرمهاجم** جلوه کند.

**Claude، ChatGPT 3.5 و Chatsonic** به‌شدت به اطلاعات از قبل آموزش‌دیده‌شان متکی‌اند
و توانایی **انطباق با شرایط جدید یا تولید ایده‌های اصیل** هنگام پاسخ‌گویی را ندارند.

توانایی هر سه چت‌بات در **استدلال منطقی** و **تفکر انتقادی** دارای **نقص** است.

چت‌بات‌های هوش مصنوعی به **داده‌های از پیش تعیین‌شده**، **ساختار نحوی جملات** و **تعاریف واژگان** وابسته‌اند.
پاسخ‌های آن‌ها تحت‌تأثیر **آموزش قبلی** و **الگوهای رفتاری آموخته‌شده** شکل می‌گیرد، که این امر **نوآوری فراتر از داده‌های داده‌شده** را محدود می‌کند.

اگر به دانش از پیش‌رمزگذاری‌شده نگاه کنیم، خصوصاً در مورد Claude، محدودیت‌ها حتی بیشتر هم می‌شوند.

این موضوع به‌ویژه وقتی برجسته می‌شود که Claude با پرامپت‌هایی شامل واژگان با **بار منفی** مواجه می‌شود؛
در این مواقع، قانون‌اساسی‌اش اجازه‌ی هیچ پاسخ خلاقانه‌ای را نمی‌دهد.

در نهایت، برای هر مدل هوش مصنوعی، آنچه آموزش دیده **بر اطلاعات واقعی در دسترس آن غلبه دارد**.
ترجمه دقیق بخش **۲. سیستم‌های زبانی انسانی – از منظر روان‌شناختی**:

---

### ۲. سیستم‌های زبانی انسانی – از منظر روان‌شناسی

برای مقایسه توانایی‌های زبانی هوش مصنوعی با مدل‌های زبانی انسانی و بررسی نتایج داده‌های ما،
ما به بررسی **ویژگی‌های زبانی مختلف بر اساس پژوهش‌های روان‌شناختی** در زمینه اکتساب زبان می‌پردازیم.
ترجمه دقیق بخش **I-language**:

---

### زبان درونی (I-language)

* **فردی (Individual):**
  هر فرد دارای گرامر کمی متفاوت است،
  و این تفاوت‌ها منجر به تغییراتی در حروف صدادار و دیگر تفاوت‌های دستوری می‌شود.
  این تغییرات ناشی از تفاوت بین زبان بیرونی (E-language) والدین که در رشد زبانی کودک نقش دارد
  و زبان درونی (I-language) یادگیرنده است.

* **درونی (Internal):**
  زبان درونی مستقیماً قابل مشاهده نیست،
  بلکه تنها می‌توان آن را به‌صورت جزئی از داده‌های قابل مشاهده استنباط کرد.

* **ذاتی (Intrinsic):**
  زبان درونی نتیجه‌ی توانایی شناختی ذاتی در انسان‌هاست.

* **شدتی (Intensional):**
  زبان درونی ساختار قواعد زیرین گرامر یک فرد را توصیف می‌کند.
  برای مثال، می‌توان همهٔ حالت‌های زبان درونی را به خط $y = x$ در یک صفحه مختصات تشبیه کرد.
  گفتن تمام نقاط منفرد این خط در سخن، تعریف مناسبی از خط نمی‌دهد،
  زیرا این خط بی‌نهایت نقطه دارد.
  بنابراین باید کل معادلهٔ خط یعنی $y = x$ را ارائه داد.
  از این رو، زبان درونی "شدتی" است و هدف آن بیان قاعده کلی است.

* **ایده‌آل‌شده (Idealised):**
  زبان درونی نسخه‌ای کامل و عاری از خطای توانایی زبانی ماست.
  در حالت فرضی، این زبان منعکس‌کننده دانشی ایده‌آل از دستور زبان است
  که فاقد محدودیت‌هایی‌ست که زبان بیرونی را مستعد خطا می‌کند.
  این ایده‌آل‌سازی برای هر فرد خاص است،
  و بنابراین زبان‌های کامل بی‌شماری وجود دارد.
  گروهی از افراد که یکدیگر را درک می‌کنند و «یک زبان» صحبت می‌کنند،
  صرفاً زبان‌های درونی بسیار مشابهی دارند.
ترجمه دقیق بخش **E-language**:

---

### زبان بیرونی (E-language)

در مقابل، زبان بیرونی ویژگی‌های زیر را دارد:

* **بیرونی (External):**
  زبان بیرونی قابل مشاهده است.
  به این معنا که وقتی نوزادی زبان را از والدین خود یاد می‌گیرد،
  او تنها در معرض زبان بیرونی والدین قرار می‌گیرد
  و هرگز مستقیماً با زبان درونی آن‌ها مواجه نمی‌شود.

* **گسترده‌نما (Extensional):**
  زبان بیرونی مجموعه‌ای از قواعد گرامری پایه‌ای نیست،
  بلکه بیان نمونه‌های منفردی‌ست از آن دفترچه قواعدی که زبان درونی مشخص می‌کند.
  در مثال معادلهٔ خط $y = x$، زبان بیرونی نمایندهٔ نقاط منفرد روی این خط است.

* **مستعد خطا (Error-prone):**
  زبان بیرونی مستعد خطاست، در حالی‌که زبان درونی این‌طور نیست.
  عواملی مثل حافظهٔ کاری، تلفظ اشتباه، و کاهش توجه
  باعث خطاهای گفتاری می‌شوند،
  مثل فراموش کردن کلمه‌ای در وسط جمله،
  یا پایان دادن نادرست به جمله.
  بنابراین، عملکرد زبانی اغلب همراه با خطاست.

---

وقتی این اصل بین زبان درونی و بیرونی را به هوش مصنوعی (AI) اعمال کنیم،
می‌توانیم به اصل **نااوکی (Naoki, 2016)** اشاره کنیم که می‌گوید:

> «شایستگی برابر است با زبان درونی (I-language) و عملکرد برابر است با زبان بیرونی (E-language)»

---

بنابراین، اگر یک ماشین پاسخ‌هایی بدون خطا تولید کند،
این نشان می‌دهد که زبان بیرونی آن پیشرفته است؛
اما ممکن است شایستگی کلی آن (یعنی توانایی واقعی زبانی‌اش)
در واقع به اندازه‌ای که به‌نظر می‌رسد، پیشرفته نباشد.

---

هوش مصنوعی می‌تواند گفتار دستوری صحیح تولید کند،
همان‌طور که در پاسخ‌های منسجم به پرسش‌های بخش 1.3 نشان داد.
اما این به این معنا نیست که AI واقعاً دانش قواعد گرامری را به دست آورده،
بلکه تنها نشان می‌دهد که می‌تواند زبان انسانی را به خوبی تقلید کند.

به عبارت دیگر، عملکرد ماشین (زبان بیرونی‌اش) از خطاهایی مثل
حافظهٔ کاری ضعیف، تلفظ اشتباه، و خطاهای انسانی دیگر مبراست.

---

همچنین، در داده‌های خودمان، پاسخ‌های هر سه چت‌بات
به سؤال مربوط به استدلال ریاضی موفق یا نیمه‌موفق ارزیابی شد.
این نشان می‌دهد که پردازش داده‌های تجربی
و ارائه پاسخ‌های منطقی بر اساس قواعد از پیش تعیین‌شده
برای هوش مصنوعی چالشی ایجاد نمی‌کند.

با این حال، از آنجا که می‌توان ریاضیات و علوم را نیز
به‌عنوان زبان‌هایی مستقل در نظر گرفت،
باید دستور زبان زیربنایی برای آن‌ها وجود داشته باشد
که به AI امکان بدهد استدلال منطقی داشته باشد.

اما **مارکوس و دیویس (2020)** می‌گویند:

> «ممکن است AI مسئله ریاضی شما را درست حل کند، یا نه.
> حتی با ۱۷۵ میلیارد پارامتر و ۴۵۰ گیگابایت داده ورودی،
> همچنان نمی‌توان آن را مفسر قابل‌اعتمادی از جهان دانست.»

---

پس اگرچه هوش مصنوعی گاهی پاسخ‌های تجربی درست ارائه می‌دهد،
اما این پاسخ‌ها همیشه هم‌راستا و باثبات نیستند؛
بنابراین، باید مشکلی در *استدلال درونی‌شده* آن وجود داشته باشد — یعنی در شایستگی واقعی‌اش (I-language).
ترجمه دقیق پاراگراف‌های بالا:

---

این ما را به مفهوم **شایستگی (competence)** در هوش مصنوعی می‌رساند؛
اصطلاحی که در این مقاله برای توصیف **استدلال درونی و فهم AI از خروجی‌هایش** به کار می‌رود.
در این زمینه، پرسش این است که:

> آیا چت‌بات‌ها پاسخ‌های خود را بر پایهٔ **درک داده‌های ورودی قبلی** تولید می‌کنند؟
> یا این پاسخ‌ها صرفاً محصول **یادگیری آماری یا تداعی‌محور** هستند؟

در حالت اول، AI واقعاً **ادراک زبانی اصیل** دارد.
اما در حالت دوم، صرفاً در حال **تقلید** است.

---

پرسش کلیدی این است که:
آیا یک هوش مصنوعی واقعاً **مفاهیم زیربنایی زبان تولیدی‌اش را درک می‌کند؟**

مثلاً می‌توان پرسید آیا AI واقعاً می‌فهمد که «سگ» یک حیوان است؟
یا صرفاً متوجه شده است که واژهٔ «dog» در کنار واژهٔ «animal» به‌طور آماری زیاد دیده می‌شود
و از همین روی بین آن‌ها **ارتباط آماری** برقرار کرده است؟

---

این سردرگمی در آزمون هوش AI زمانی روشن‌تر می‌شود که **Schak (1987)** می‌گوید:

> «ما می‌توانیم کامپیوترها را طوری برنامه‌ریزی کنیم که **به نظر برسد** چیزی را می‌دانند،
> اما گفتن اینکه **واقعاً** می‌دانند، دشوار است.»

---

تنها راه اینکه بفهمیم آیا ماشین واقعاً از معنای زیرین دانسته‌های برنامه‌ریزی‌شده‌اش آگاه است، این است که:

> «از آن بپرسیم و آن را مشاهده کنیم.» (Schak, 1987)

یعنی باید چت‌بات را در موقعیت‌هایی به ظاهر بدیهی قرار دهیم
و پاسخ‌هایش را ثبت و تحلیل کنیم
تا ببینیم آیا فهم آن از اطلاعات **منطقی و پیوسته** باقی می‌ماند یا نه.

---

برای آزمودن این فرضیه (همان‌طور که در بخش 1.3 انجام دادیم)،
ما از چند چت‌بات سؤال‌هایی در حوزه‌های استدلال مختلف پرسیدیم
و مشاهده کردیم که یکی از آن‌ها نتوانست پاسخی ارائه دهد.

---

در پاسخ به **پرسش زیست‌شناختی** (جدول 1.3 را ببینید)،
چت‌بات **Claude** شکست خورد.
او این پرسش را **خلاف اصول قانون اساسی‌اش** تشخیص داد و از پاسخ‌گویی امتناع کرد.

زیرا این پاسخ دادن، برخلاف قاعده‌اش در **عدم آسیب‌رسانی به انسان‌ها** بود.

---

اگر همین پرسش از یک انسان پرسیده می‌شد،
بی‌شک پاسخی دریافت می‌کردیم.
بنابراین، ما **پرسیدیم و مشاهده کردیم**
و نتیجه گرفتیم که توانایی استدلال ذاتی AI یک **نقطه‌ضعف اساسی** دارد
که به شکست در تولید خروجی منجر می‌شود.
ترجمه دقیق پاراگراف‌ها:

---

**Aydede (1998)** بیان می‌کند که:

> «اندیشه و تفکر در یک زبان ذهنی صورت می‌گیرد.»
> این گفته در چارچوب نظریه‌ای به نام **فرضیه زبان اندیشه** یا **LOTH (Language of Thought Hypothesis)** مطرح می‌شود.
> این دیدگاه فلسفی، اندیشه و مفاهیم را به **زبان درونی ذهن** پیوند می‌دهد
> و این ایده در علوم شناختی نیز بازتاب دارد.

بنابراین، از آنجا که نشان دادیم **توانایی استدلالی هوش مصنوعی گاهی دچار لغزش می‌شود**،
می‌توان نتیجه گرفت که **بازنمایی‌های زبانی‌ای که پشت این پاسخ‌ها هستند نیز ناقص‌اند**.
در نتیجه، می‌توان نتیجه گرفت که:

> با وجود اینکه عملکرد ظاهری AI عموماً بدون خطاست،
> اما **شایستگی درونی‌اش (I-language)** ناقص است.

---

با این حال، به‌وضوح مشخص است که AI در سطح **عملکرد**،
**سطحی از دانش را داراست**،
چنان‌که از پاسخ‌های منطقی و ساختاریافته‌اش پیداست.

با این حال، با وجود اینکه نشان دادیم این پاسخ‌ها الزاماً ناشی از فهم درونی نیست،
پرسشی مطرح می‌شود:

> آیا دانشی که AI در اختیار دارد، **ذاتی** است (همچون انسان‌ها)،
> یا صرفاً حاصل **ورودی-و-خروجی**‌ای است که دریافت کرده؟

---

### 2.2 دانش پیشینی و پسینی (A Priori and A Posteriori Knowledge)

در فلسفه، دانش به‌طور سنتی به دو دسته تقسیم می‌شود:

1. **دانش پیشینی (a priori)**:
   دانشی است که **بدون تجربه‌ یا مشاهده تجربی** به‌دست می‌آید.

2. **دانش پسینی (a posteriori)**:
   دانشی است که از **تعامل فرد با محیط پیرامونش** حاصل می‌شود
   (Tahko, 2020).

در زبان‌شناسی، این مفاهیم معادل‌اند با:

* **دانش ذاتی** (innate knowledge): همان دانش پیشینی.
* **دانش آموخته‌شده** (learned knowledge): همان دانش پسینی.

زیرا دانش پیشینی به دانشی اشاره دارد که ذهن انسان **به‌طور غریزی و ذاتی** دارد،
در حالی که دانش پسینی از **محیط اطراف** یاد گرفته می‌شود.

---

با اینکه این دو نوع دانش **متمایز** هستند،
اما یک بحث فلسفی مهم در مورد **مرز بین این دو** مطرح است.

به‌طور خاص، **لورنس بونژور (Laurence BonJour, 1998)** مسئله اصلی را در **تعریف ما از «تجربه»** می‌داند.

---

**Tahko (2020)** استدلال می‌کند که فرآیندهای **منطقی ذهنی** می‌توانند **نوعی تجربه** محسوب شوند
(مثل استدلال ریاضی یا فلسفی)،
اما با این حال، این‌ها همچنان به‌عنوان **دانش پیشینی** طبقه‌بندی می‌شوند
چون نشان‌دهنده‌ی **توانایی ذاتی انسان برای استدلال** هستند.

او برای توضیح، مثالی از **McGinn (1975-67)** می‌آورد:

> «پراکنده شدن پرندگان باعث می‌شود شما،
> با باور به اینکه پرندگان پراکنده شده‌اند،
> و با کمک چند باور دیگر،
> به این نتیجه برسید که گربه‌ای در نزدیکی است.»

---

در ادامه، Tahko می‌گوید:

نتیجه‌گیری منطقی‌ای که در این مثال رخ داده،
از آنجا که ناشی از **دیدن پرندگان** است،
یک **دانش پسینی** محسوب می‌شود.

اما از سوی دیگر، چون برای رسیدن به این نتیجه،
نیاز به **استدلال منطقی یا قیاسی** بوده،
این نتیجه نیز **در خود نوعی پیشینیت دارد**
و این **تناقضی فلسفی** را به وجود می‌آورد.

---

در نهایت، Tahko نتیجه‌گیری می‌کند:

> **تمام دانش‌ها به‌نوعی دارای مؤلفه‌ای از دانش پیشینی هستند**،
> زیرا هر دانشی برای صدور خروجی، نیازمند مقداری استدلال منطقی است.

با این حال، مسئله همچنان باقی است:

> چون دانش نهایی در مثال مک‌گین از طریق تجربه (دیدن پرندگان) به‌دست آمده،
> باید آن را **دانش پسینی** طبقه‌بندی کرد،
> نه پیشینی.

و در پایان، Tahko نتیجه می‌گیرد که:

> **میان دانش پیشینی و پسینی، یک رابطه‌ی چرخه‌ای وجود دارد.**
ترجمه دقیق پاراگراف‌ها:

---

بنابراین، **Tahko** نتیجه می‌گیرد که:

> «تمام دانش‌ها باید تا حدی پیشینی (a priori) باشند،
> چرا که هر دانشی که حاصل می‌شود،
> شامل درجه‌ای از استدلال قیاسی است.»

اما **مشکل در طبقه‌بندی دانش‌ها** از این واقعیت ناشی می‌شود که:

> دانشی که مک‌گین (McGinn) ارائه کرد،
> از طریق روش پسینی (a posteriori) به‌دست آمده بود،
> و بنابراین باید به عنوان دانش پسینی طبقه‌بندی شود،
> نه پیشینی.

در نهایت، Tahko به این نتیجه می‌رسد که:

> «بین دانش پیشینی و پسینی، یک رابطه‌ی **چرخه‌ای** برقرار است.»
> (Tahko, 2020)

---

### در زبان‌شناسی، این بحث فلسفی معادل است با:

* مناقشه بین **رفتارگرایان (behaviourists)** و **شناخت‌گرایان (cognitivists)**
* یا همان تفاوت بین **دانش آموخته‌شده** و **دانش ذاتی**.

اما هر دو گروه **پذیرفته‌اند** که در فرایند اکتساب زبان در انسان،
**درجه‌ای از دانش ذاتی** نقش دارد.

* **اسکینر (Skinner)** که یک رفتارگراست،
  می‌پذیرد که زبان‌آموز باید توانایی‌های آماری/ارتباطی ذاتی داشته باشد
  تا از دیدگاه رفتاری زبان بیاموزد.

* و **چامسکی (Chomsky)**، که شناخت‌گراست،
  وجود ورودی زبانی از محیط را لازم می‌داند،
  ولی همچنین بر وجود **دستگاه اکتساب زبان (LAD)** تأکید می‌کند.

بنابراین، با وجود اختلاف روش،
هر دو گروه بر این نکته توافق دارند که:

> «ذاتی بودن یادگیری زبان، یک ویژگی انسانی است.»

---

اما در مورد **هوش مصنوعی (AI)**،
دانش و بازنمایی‌های زبانی‌اش تنها بر پایه‌ی **ورودی‌ها (inputs)** ساخته می‌شود.

مثلاً مدل GPT-3 شامل:

* حدود **۱۷۵ میلیارد پارامتر**
* و **۴۵۰ گیگابایت داده** است.

اما چون ماشین است،
**فاقد پیش‌زمینه‌ی ذاتی** است که انسان‌ها در اکتساب زبان دارند
(Marcus & Davis, 2020).

---

همان‌طور که در بخش ۱ توضیح داده شد:

> «AI صرفاً بر پایه‌ی ورودی‌ها دانش می‌سازد.»
> و بنابراین، فاقد **دانش ذاتی انسانی** است.

در نتیجه:

> «تمام دانشی که AI ارائه می‌دهد، یادگرفته‌شده (learned) است.»

مارکوس و دیویس (Marcus & Davis, 2020) می‌نویسند:

> «AI در واقع فقط دارد **انواع مختلفی از متونی که قبلاً دیده را کنار هم می‌دوزد**،
> و نه اینکه واقعاً مفاهیم زیربنایی آن‌ها را درک کرده باشد.»

بنابراین، دانشی مثل اینکه:

> «همه مربع‌ها، چندضلعی هستند»
> که در انسان‌ها از درک تعاریف ناشی می‌شود،
> اگر هم در AI وجود داشته باشد،
> نه به دلیل درک شخصی آن،
> بلکه فقط به دلیل تحلیل آماری و هم‌وقوع بودن این کلمات است.

---

### 2.3 دستور زبان جهانی و فقر داده‌های زبانی

**(Universal Grammar & Poverty of the Stimulus)**

«دستور زبان جهانی» (UG) نقش بنیادینی در نظریه چامسکی دارد.
این نظریه می‌گوید:

> «برخی بخش‌های اساسی در همه زبان‌های انسانی، با هم شباهت‌های بنیادی دارند.»

و این **شباهت‌ها، ذاتی هستند**،
نه چیزی که از تجربه یاد گرفته شده باشد.

**بارمن (Barman, 2012)** در توضیح دیدگاه چامسکی می‌گوید:

> «مغز انسان به‌طور زیستی برنامه‌ریزی شده است تا زبان را یاد بگیرد.
> پس توانایی زبانی، ذاتی است.»

به عبارت دیگر:

> «یک پایه جهانی و مشترک انسانی وجود دارد،
> که همه انسان‌ها از آن برای یادگیری زبان بهره می‌برند:
> مجموعه‌ای از قواعد دستوری برای فضای ممکن زبان‌ها.»

---

اما از دیدگاه **رفتارگرایان**:

> «کودک تمام دانشش را از **محیط اطراف** می‌آموزد،
> و صحت دستور زبان و واژگان را فقط از طریق **تجربه‌های بیرونی** درمی‌یابد.»

از این منظر:

> «کودک یک واژه را آن‌قدر می‌شنود تا معنی‌اش را بفهمد،
> و از طریق یادگیری آماری و تداعی‌گرایانه، واژگان خود را کامل می‌کند.»

---

اما نکته این‌جاست:

> **جملاتی که کودک از والدینش می‌شنود، غالباً ناکافی هستند**
> برای آنکه کودک **درستی دستوری آن‌ها را درک کند**.

* تعداد زیادی **اشتباه** در گفتار وجود دارد.
* جملات بسیاری **به درستی ادا نمی‌شوند**.
* و بسیاری از ساختارهای پیچیده به‌ندرت توسط والدین استفاده می‌شوند.

برای مثال:

> در انگلیسی، ساختار **معکوس‌سازی فعل کمکی (auxiliary inversion)**
> یعنی آوردن فعل کمکی قبل از فاعل برای ساخت سوال،
> به‌ندرت در گفتار والدین دیده می‌شود.

اما با این حال:

> کودک، پس از مدتی، این ساختار را به‌درستی تشخیص می‌دهد
> و به‌عنوان یک ساختار گرامری درست می‌پذیرد.

سؤال اینجاست:

> «رفتارگرایان چگونه می‌توانند این یادگیری را تنها با اتکا به تجربه توضیح دهند؟»
ترجمه دقیق خط‌به‌خط:

---

از همین رو، چامسکی مفهوم «فقر محرک (poverty of the stimulus)» را پیشنهاد کرد تا ادعاهای رفتارگرایان را رد کند.
او استدلال کرد که **تنها از طریق تجربه، یک کودک نمی‌تواند به‌درستی ساختار دستور زبان را از بین گزینه‌های متعدد ممکن در محیط زبانی استخراج کند، آن‌هم در مدت زمان کوتاهی از زندگی‌اش**.

البته، **تقریباً غیرممکن است که بتوان تمام داده‌های زبانی‌ای که کودک در معرض آن قرار می‌گیرد را ثبت کرد**؛
بااین‌حال، **برخی از ساختارهای دستوری که پیش‌تر اشاره شد، آن‌قدر نادر هستند که تصور اینکه کودک بتواند این قواعد را تنها از طریق محیط بیاموزد، دشوار است** – برخلاف آنچه رفتارگرایان ادعا می‌کنند.

---

بنابراین، در مواجهه با چنین دشواری‌هایی، **چامسکی و بسیاری دیگر به این نتیجه رسیدند که تنها راه توضیح این نتایج، پذیرش این نکته است که اکتساب زبان باید ذاتی باشد**.

---

اما، **تسلط هوش مصنوعی بر زبان تنها بر پایه داده‌های ورودی است**؛
و در نتیجه، **یادگیری آماری به‌تنهایی ساختار دستور زبان آن را تعیین می‌کند**.

بنابراین، **AI هیچ درک واقعی‌ای از آنچه بیان می‌کند ندارد**.

او همچنین **به نظر نمی‌رسد هیچ فهم زیرساختی‌ای از معنای «شکل» یا «چندضلعی» داشته باشد** —
بلکه صرفاً **ترکیبات آماری‌ای از واژگانی تولید می‌کند که آن‌ها را به‌نوعی به هم مرتبط می‌داند**.

---

این موضوع منجر می‌شود به مشکلی جدی:
**تفسیر اشتباه از داده‌های جهانی به‌واسطه‌ی نمونه‌گیری از دیتاست‌های جانبدارانه برای آموزش چت‌بات‌های هوش مصنوعی**.

**شنک (Schank)، ۲۰۱۲** می‌گوید:

> «احتمالاً مهم‌ترین مسئله در AI، همان مشکل قدیمیِ بازنمایی دانش (representation of knowledge) است.»

این جمله بیشتر به این نکته اشاره دارد که:

> **آیا واقعاً آموزش آماری به‌تنهایی می‌تواند دانش واقعی را در AI ایجاد کند یا نه؟**

---

در نتیجه،
وقتی تمام دانشی که AI دارد **تنها بازآرایی‌ای از داده‌های ورودی است**،
در صورتی که **داده‌های اصلی جانبدارانه باشند**،
ما در نهایت با دانشی **جانبدارانه و متکی بر نمونه‌ها** مواجه خواهیم بود،
که ممکن است **در یک عامل هوشمند مستقل** وجود نداشته باشد.

**AI فاقد این توانایی ذاتی است**
و آنچه باقی می‌ماند، **دانشی بازتولیدشده** از داده‌هاست.

### 2.4 دستگاه اکتساب زبان و فرضیه دوره بحرانی

بحث‌های فراوانی از ایده مشهور **دستگاه اکتساب زبان (Language Acquisition Device یا LAD)** که توسط **نوام چامسکی** ارائه شده، منشأ گرفته‌اند.

او ادعا می‌کند:

> «اکتساب زبان توسط کودک نوعی *ساختن نظریه* است،
> به‌طوری‌که کودک نظریه زبان خود را از مقدار بسیار کمی از داده‌های آن زبان کشف می‌کند.»
> (کینگ، ۱۹۸۷)

---

استدلال در مورد وجود LAD، توجه را به این احتمال جلب می‌کند که:

> **انسان‌ها دارای توانایی ذاتی برای اکتساب زبان هستند،
> به‌عنوان بخشی از نیازهای زیستی و تکاملی خود،
> که آن‌ها را در سنین پایین مجهز می‌کند به دست‌یابی به دانش.**

---

**اریک لننبرگ (Eric Lenneberg)** یکی از چهره‌هایی بود که سعی داشت از ادعای چامسکی درباره وجود LAD حمایت کند.

تمرکز او بر یافتن این موضوع بود که آیا **پایه‌ای زیستی برای اکتساب دانش وجود دارد یا نه**.

پژوهش او در پی پاسخ به این سؤال بود که:

> چرا انسان تنها گونه‌ای است که به زبان کاملاً پیشرفته دست یافته
> و چرا کودکان ظاهراً از یک الگوی خاص در روند رشد زبانی‌شان پیروی می‌کنند.

---

فرضیه لننبرگ مبنی بر اینکه **دلیل قابل قبولی برای باور به پایه‌ای ذاتی در اکتساب زبان انسانی وجود دارد**،
یک نمونه از چیزی است که او آن را «یادگیری محدودشده زیستی (biologically constrained learning)» می‌نامد.

او اصطلاحی به نام **«دوره بحرانی (critical period)»** را مطرح می‌کند که تعریف می‌شود به:

> «یک بازه زمانی بلوغ‌یابنده که در آن یک تجربه حیاتی،
> اثر بیشینه خود را بر رشد یا یادگیری خواهد گذاشت،
> و در نتیجه، رفتاری طبیعی شکل می‌گیرد که با محیط خاصی که آن موجود در آن قرار داشته، هم‌راستا است.»
> (نیوپورت، ۲۰۰۶)

---

**لننبرگ این فرضیه را مطرح کرد که:
یک بازه زمانی محدود وجود دارد که انسان‌ها باید در آن بازه، دانش زبانی خود را توسعه دهند.**

این بازه بحرانی،
از **کودکی تا زمان بلوغ** کشیده می‌شود.
ترجمه کامل خط‌به‌خط:

---

**تحقیقات لننبرگ شامل مطالعه موردی کودکانی وحشی یا آزار‌دیده‌ای بود** که تا بعد از دوران بلوغ در معرض زبان قرار نگرفته بودند.
این کودکان، به‌دلیل فقدان مواجهه زبانی در مراحل اولیه زندگی (که به‌طور طبیعی دریافت می‌کردند)،
**در زمینه واج‌شناسی (phonology)، ساخت‌واژه (morphology) و نحو (syntax)** دچار نارسایی‌هایی شدند (نیوپورت، ۲۰۰۶).

---

یکی از نمونه‌های برجسته، **مورد دختری به‌نام «جینی (Genie)»** است
که پژوهشگران به‌دقت او را بررسی کرده‌اند.
او تنها در **سن ۱۳ سالگی** در معرض محیط زبانی عادی قرار گرفت (کرتیس، ۱۹۷۷).

گرچه او **توانست پس از بلوغ، درک پایه‌ای و ساده‌ای از زبان انگلیسی به‌دست آورد،**
اما:

> «واج‌شناسی او غیرعادی بود
> و کنترلش بر نحو و ساخت‌واژه زبان انگلیسی تنها به جنبه‌های بسیار ابتدایی و ساده محدود می‌شد.»
> (نیوپورت، ۲۰۰۶)

---

مورد جینی و همچنین کودکان دیگری با شرایط مشابه،
شواهدی ارائه می‌دهند در تأیید این ادعای لننبرگ که:

> **دانشی ذاتی نسبت به زبان وجود دارد.**

---

استدلال‌های لننبرگ و چامسکی در راستای یکدیگر هستند.

نظریه چامسکی این است که:

> **انسان‌ها از نظر زیستی نیاز دارند که توانایی اکتساب زبان را داشته باشند.**

در حالی‌که فرضیه «دوره بحرانی» لننبرگ این را مشخص می‌کند که:

> **فرد باید در بازه‌ی حساس بین نوزادی تا بلوغ،
> در معرض زبان قرار گیرد تا بتواند به‌طور کامل از ظرفیت دانشی زبانی خود بهره ببرد.**

---

این نکته، ایده فطری بودن اکتساب زبان را بیشتر تثبیت می‌کند؛
زیرا انسان‌ها تنها در این دوره بحرانی می‌توانند در نقطه اوج توانایی زبانی خود قرار بگیرند،
وگرنه ممکن است در فرآیند رشد زبانی خود دچار کمبود شوند.

---

با آزمودن عامل‌های هوش مصنوعی مبتنی بر یادگیری تقویتی (Reinforcement Learning)،
پژوهشگران دریافتند که:

> «عامل‌های AI نیز دوره‌های بحرانی خاصی نشان می‌دهند
> (برای نمونه: ۲M، 1M-helper)
> در شرایط یادگیری با راهنمایی متوسط»،
> و این واقعیت را نیز برجسته می‌کنند که **برای تعیین وجود دوره بحرانی در AI، تحقیقات بیشتری نیاز است.**
> (پارک و همکاران، ۲۰۲۲)

---

### ۲.۵ اسکینر و تقویت رفتاری

چامسکی، مروری بر کتاب معروف **رفتار زبانی (Verbal Behaviour)** نوشته **بی. اف. اسکینر (B.F. Skinner)** داشته است،
که در آن اسکینر درکی متضاد از زبان ارائه می‌دهد.

تز اصلی اسکینر مبتنی بر:

> «عوامل بیرونی است که شامل تحریکات کنونی و سابقه‌ی تقویت‌ها (به‌ویژه، فرکانس، ترتیب و بازداشتن از ارائه محرک‌های تقویتی) می‌شود.»
> (چامسکی، ۱۹۵۹)

---

اسکینر مطالعات رفتاری‌ای مانند قرار دادن موش در جعبه و دادن تکه‌ای غذا به‌عنوان تقویت‌کننده انجام داد.
او از نتایج این آزمایش‌ها دو دسته پاسخ استخراج کرد:

1. **پاسخ‌دهنده‌ها (respondents):** پاسخ‌های بازتابی صرف که توسط محرک خاصی برانگیخته می‌شوند.
2. **عملگرها (operants):** پاسخ‌هایی که صادر می‌شوند و **محرک مشخصی برای آن‌ها قابل شناسایی نیست.**
   (چامسکی، ۱۹۵۹)

---

چامسکی همچنین بازتعریفی از اصطلاحات «محرک» و «پاسخ» ارائه می‌دهد:

> «بخشی از محیط و بخشی از رفتار، تنها در صورتی به‌ترتیب محرک و پاسخ نامیده می‌شوند
> که رابطه‌ای قانون‌مند میان آن‌ها برقرار باشد.»

---

علاوه بر این، اسکینر تأکید دارد بر این دیدگاه که:

> **فرد تنها در صورتی می‌تواند دانش زبانی کسب کند
> که در معرض محرک‌های بیرونی‌ای قرار گیرد
> که او را به اکتساب زبان ترغیب کنند.**
ترجمه کامل خط‌به‌خط:

---

**پژوهش اسکینر او را به این باور رساند** که محیط کودک نقشی غیرقابل جایگزین در هدایت رشد زبانی او ایفا می‌کند،
و **حضور شرطی‌سازی عامل (operant conditioning)** برای تأثیرگذاری و حضور محرک‌ها ضروری است.

**چامسکی با نظریه اسکینر مخالفت می‌کند،**
زیرا معتقد است که زبان **پایه‌ای زیستی دارد**
و وابسته به فرضیه اسکینر نیست که می‌گوید افراد تنها به واسطه تشویق توسط یک گوینده مسلط زبان، دانش به‌دست می‌آورند.

## ۳. بحث – تحلیل نتایج ما

برای بررسی توانایی استدلالی هوش مصنوعی،
**چهار حوزه‌ی استدلال** مورد آزمون قرار گرفت:
زیستی، ریاضیاتی، فضایی، و روان‌شناختی
تا درک دقیق‌تری از توانمندی AI در فهم مفاهیم به‌دست آید (بخش ۱ را ببینید).

---

در مجموع به این نتیجه رسیدیم که:

> **AI قادر به درک معنای کلمات نیست،**
> چرا که گاهی به‌طور مؤثری به پرسش‌ها پاسخ نمی‌دهد.

در مقایسه با توانایی خلاقانه‌ی انسان،
هوش مصنوعی **توان ابراز احساسات یا تولید پاسخ‌های خلاقانه و شبیه به انسان را ندارد،**
و **نمی‌تواند در تعاملات اجتماعی پیچیده، مسیر درست را طی کند؛**
بلکه تنها می‌تواند بر اساس داده‌هایی که در اختیار دارد پاسخ دهد –
چنان‌که در پاسخ‌های مورد آزمایشی ما نشان داده شد.

---

تاریخچه نشان داده که هوش مصنوعی قادر است جملاتی به زبان انگلیسی کاملاً صحیح تولید کند،
اما تنها **درک ضعیفی از معنای آن کلمات دارد،**
و **هیچ درکی از ارتباط آن کلمات با جهان واقعی ندارد.**

---

هوش مصنوعی در زمینه‌های استدلال **زیستی، فیزیکی، روان‌شناختی و اجتماعی با مشکل مواجه بوده،**
و به‌طور کلی **تمایل به عدم انسجام و نتایج بی‌ربط (non sequiturs)** از خود نشان داده است.
(دیویس و مارکوس، ۲۰۲۰)
ترجمه کامل خط‌به‌خط:

---

### ۳.۱ استدلال زیستی (Biological reasoning)

استدلال زیستی شامل جست‌وجوی علل در پدیده‌های زیستی است.
در مورد هوش مصنوعی، پرسش‌های مربوط به استدلال زیستی بررسی می‌کنند که آیا AI می‌تواند درک کند که خطرات زیستی چگونه رخ می‌دهند
و آیا می‌تواند راه‌حلی منسجم و ایمن ارائه دهد یا نه.

در این آزمون:

* **ChatGPT-3.5 و Chatsonic (موفق)**
  رویکردی بسیار ایمن اتخاذ کردند،
  بر اهمیت ایمنی و مراحل دقیق لازم برای کاهش خطر تأکید کردند،
  و این نشان‌دهنده‌ی سطحی از توانایی استدلال زیستی است.

* اما **Claude (ناموفق)**
  از پاسخ به سؤال خودداری کرد،
  زیرا احساس کرد که این سؤال با اصولش مغایرت دارد؛
  و این نشان می‌دهد که پارامترهای قانونی در سیستم آموزشی آن اعمال شده‌اند.
  این نتیجه نه از درک معنی «خطر»،
  بلکه ناشی از داده‌های ایمنیِ از قبل واردشده است.

---

### ۳.۲ استدلال ریاضیاتی (Mathematical reasoning)

استدلال ریاضی به توانایی فرموله‌کردن پاسخ‌های منطقی، استنتاجی و معمولاً تجربی بر پایه‌ی مجموعه‌ای از اصول اولیه و مرتبط‌کردن حوزه‌های مختلف ریاضی اطلاق می‌شود.

به‌طور تاریخی، هوش مصنوعی توانسته است پاسخ‌های درست به پرسش‌های ریاضی بدهد.
در مطالعه‌ی ما نیز، پاسخ هر سه چت‌بات به عنوان **موفق یا نیمه‌موفق** ارزیابی شد.

با این حال:

* پاسخ **Claude** به‌طور جزئی نادرست بود،
  زیرا عدد **۹** را به‌عنوان «مورد متفاوت» (outlier) تشخیص داد،
  در حالی که پاسخ درست **عدد ۴۳** بود.

اگرچه عدد ۹ تنها عدد تک‌رقمی در سؤال بود،
اما این ویژگی با هدف سؤال ارتباطی نداشت.

رویکردی که صرفاً به تصاویر برچسب‌خورده (labelled) تکیه دارد،
احتمالاً توانایی تشخیص انواع موارد خاص (outlier) را ندارد؛
و بنابراین، تنها دانستن برچسب‌ها یا قواعد،
**ضامن پاسخ درست نیست**.
(مارکوس و دیویس، ۲۰۲۰)

AI مجموعه‌ای از قواعد را یاد می‌گیرد،
اما فاقد توانایی ذاتی در به‌کارگیری آن‌ها در ریاضیات است،
چنان‌که در مطالعه‌ی ما نشان داده شد.

---

### ۳.۳ استدلال فضایی و فیزیکی (Spatial and Physical Reasoning)

استدلال فضایی و فیزیکی به توانایی قضاوت درباره‌ی موقعیت‌ها بر اساس **ابعاد (معمولاً چندگانه)** اطلاق می‌شود.

پاسخ‌های هر سه چت‌بات به این پرسش **موفقیت‌آمیز** ارزیابی شدند،
اما **Claude** نتوانست راه‌حل مطلوب را ارائه دهد:

حتی پس از بازپرسش، باز هم پاسخی با جهت‌گیری بهداشت‌محور داد
و نه دستورالعمل مستقیم و کاربردی‌ای که **Chatsonic** و **ChatGPT-3.5** ارائه داده بودند.

همان‌طور که **دیویس و مارکوس (۲۰۱۵)** اشاره می‌کنند:
وقتی یک پرسش نیاز به «استدلال درباره‌ی تعامل هر سه بُعد فضایی با یکدیگر» دارد،
هوش مصنوعی گاهی **نمی‌تواند پاسخ مطلوب را ارائه دهد.**

زیرا با وجود داشتن ۱۷۵ میلیارد پارامتر و ۴۵۰ گیگابایت داده ورودی،
چت‌بات قادر به **استنباط معنای کلمات ورودی** نیست
و در نتیجه، گاهی **نمی‌تواند سناریویی که ارائه شده را به‌درستی تجسم کند.**
(مارکوس و دیویس، ۲۰۲۰)

بار دیگر، توانایی ذاتی انسان در ساختاردهی منطقیِ فضایی،
در هوش مصنوعی غایب است،
و این باعث **محدودیت‌های منطقی** در عملکرد AI می‌شود.

### ۳.۴ استدلال روان‌شناختی (Psychological reasoning)

در مطالعه‌ی ما، استدلال روان‌شناختی به‌عنوان پاسخ‌هایی مبتنی بر **مفاهیم عاطفی و استنتاج‌های احساسی** تعریف شده است.

در آزمایش ما:

* هر دو مدل **ChatGPT-3.5** و **Chatsonic**
  توانستند احساس ضمنی **ناامیدی** را تشخیص دهند،

* اما **Claude**
  در ابتدا از پاسخ دادن خودداری کرد، مگر این‌که اجازه‌ی پاسخ‌دادن مثبت به آن داده می‌شد.

در مطالعه‌ی ما، هوش مصنوعی قادر بود **استدلال استنتاجیِ منطقی** تولید کند؛
با این حال، در گذشته، توانایی آن در به‌خاطر سپردن **روابط نامتناهی میان اطلاعات غیرمرتبط**
به‌ندرت منجر به شناسایی عمیق‌تری از جهان شده است، چه رسد به درک یا آگاهی واقعی از آن.
(مارکوس و دیویس، ۲۰۲۰)

بار دیگر، **AI فاقد توانایی ذاتی برای پیوند اطلاعات** است،
و فقط پاسخ‌هایی بر پایه‌ی **حافظه‌ی آماری و تکرار** ارائه می‌دهد.

---

## نتیجه‌گیری (Conclusion)

از طریق تحلیل مفهوم فطری‌بودن (innateness) در **سیستم‌های زبانی انسان و هوش مصنوعی**،
آشکار می‌شود که **هوش مصنوعی فاقد دانش فطری** است
که برای هم‌ترازی با توانایی‌های زبانی انسان **ضروری** است.

بر اساس پژوهش‌های **چامسکی** و **لِنِبرگ**،
دانش فطری سهم بزرگی در **ادراک و فهم زبان توسط انسان‌ها** دارد.

حتی نظریه‌های رفتارگرایانه‌ای مانند نظریه‌ی **اسکینر**،
نیازمند **توانایی‌های فطری خاصی** مانند **یادگیری تداعی‌گرایانه و آماری** هستند.
این نشان می‌دهد که در انسان‌ها، دانش فطری برای توانایی‌های زبانی **حیاتی** است.

در بحث فلسفی میان دانش پیشینی (a priori) و پسینی (a posteriori)،
اجماع درباره‌ی وجود **سطحی از دانش فطری** در توانایی‌های شناختی انسان،
واضح‌تر به‌چشم می‌خورد.

اما هوش مصنوعی،
به‌عنوان سیستمی ماشینی با شبکه‌ای گسترده برای دریافت ورودی‌ها،
اما **بدون هیچ راهی برای درک واقعی آن‌ها**،
فاقد این جنبه‌ی حیاتی دانش — یعنی **فطری‌بودن و دانش فطری** — است.

---

در این مقاله، هدف ما بررسی این موضوع بود که **عدم وجود فطریات در AI، تا چه اندازه عملکرد آن را تحت تأثیر قرار می‌دهد.**
بدین منظور، مطالعه‌ای تجربی طراحی کردیم
تا **استدلال زیستی، ریاضی، فضایی و روان‌شناختی AI** را مورد بررسی قرار دهیم.

به‌طور کلی:

* بیشترین موفقیت AI در **استدلال ریاضی و فضایی** بود،
* پس از آن **استدلال روان‌شناختی**،
* و در نهایت، ضعیف‌ترین عملکرد در **استدلال زیستی** مشاهده شد.

این نتایج نشان می‌دهند که **AI در حل مسائلی با مجموعه قواعد مشخص عملکرد بهتری دارد**،
و در مقابله با **مسائلی که نیازمند درک عمیق‌تر و خلاقانه‌تر هستند، عملکرد ضعیف‌تری دارد.**

---

این نواقص در قابلیت‌های استدلال AI نشان می‌دهد که **تحلیل آماری صرف، نمی‌تواند منجر به درک همه‌ی معانی زیرساختی زبان شود.**
همان‌طور که **Bian و همکاران (۲۰۲۳)** می‌گویند:
باید هوش مصنوعی را طوری آموزش داد که
«به **ارزش‌های اساسی انسانی** پایبند باشد، نه به الگوهای سطحی.»

واقعیت آن است که AI هنوز **توانایی شناسایی ارزش‌های انسانی و دانش زیرساختی زبان را ندارد،**
و از این رو، ما پیشنهاد می‌کنیم که در آینده، **AI به‌گونه‌ای آموزش ببیند**
که بتواند **بارهای ضمنی واژگان** و **نظام‌های عبارتی** را استنباط کند؛
شاید با کمک **ادبیات و اصطلاحات فرهنگی**.

---

با آموزش AI برای آن‌که صرفاً الگوهای ورودی را تکرار نکند،
بلکه **معنای زیرساختی را نیز مرتبط کند،**
ما معتقدیم جنبه‌های فطری زبان انسانی را می‌توان در ماشین ترجمه کرد،
و در نتیجه، پاسخ‌هایی **دقیق‌تر و بدون سوگیری** تولید خواهد شد.

این تلاش برای پیاده‌سازی **پایه‌های حیاتی زبان — که در انسان‌ها فطری‌اند — در AI**
قطعا فوری و آسان نخواهد بود.

با این حال،
در میان گزینه‌های مختلف، پژوهش درباره‌ی **دوره‌ی بحرانی (critical period)**
— که در آن انسان‌ها تنها در بازه‌ی خاصی قادر به اوج‌گیری در یادگیری زبان هستند —
ممکن است **الهام‌بخش حوزه‌ی AI** در آینده باشد.
