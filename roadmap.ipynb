بسم الله الرحمن الرحیم 

مسیری که پیش میرویم


ما برای اینکه بتوانیم هوش مصنوعی دیگری درست کنیم 
رفتیم سراغ هوش مصنوعی فعلی 
مدل های بیوهوشی مصنوعی و مدلهای مغز کامپیوتر و مدلهای دیگر رابررسی کردیم 
که در همین فایل ها وجود دارد 

بعد ما میخواستیم پردازش این همه سنگین نباشد و به صورت که انسان هم میتواند انجام شود 
یعنی مثل کودکی که اموزش میبنید و یاد میگیرد باشد ولی خب بشود از ان بهتر استفاده کرد 

در حقیقت مشکلات هوش مصنوعی فعلی را نداشته باشد 
ضمن اینکه بتوانیم در ان پیش گام هم باشیم 

به همین دلیل به برداشت های قران به خوداگاه و به حرکت جوهری نگاه کردیم 

چون دقیقا ما نمیدانیم که چیست ان چیزی که ما با ان به اگاهی میرسیم عقل است یا فطرت یا خود مغز که یکسری حرکات و حافظه و تجریه ذهنی و غیره دارد 

به همین دلیل خواستیم ببینمی اگر با حرکت جوهری یک ماده ای تولید کنیم میشود مثل قابلیت مغز داشته باشد 

که البته میرود در بیوهوش 

بعد دیدیم اگر بتوانیم بفهمانیم و یک سیستمی طراحی کنیم که بتواند خارج از شبکه های عصبی بتواند مثلا دارای هسته ای باشد که در ان یک سری بدیهیات انسان باشد 

یعنی مثلا هسته دارای منطق ارسطوی باشد 

که البته جنگ تحمیلی ایران و اسراییل شد 
و کلا رشته افکار پاره به همین دلیلی خواستیم همه چیز را بنویسیم و خط فکری را درست کنیم 

البته فایل ها موجود است 

مدل چامسکی مدل زبان جهانی مدل یکپارچگی در خوداگاهی و فهم برای ایجاد 
مدل حرکت جوهری 
و منطق ارسطویی و بدیهیات 

اینها چیزهایی هست که سعی شده فکر شودو درنظر گرفته شود 

دوم تیرماه ۱۴۰۴ چند روز قبل از محرم

فکر جدید 


میشود به این موضوع هم فکر کرد که هوش مصنوعی بر اساس قران وروایات و فلسفه اسلامی ساخت 

نه به کارگیری این هوش مصنوعی 
بلکه ساخت با هسته ای که این موضوعات را بفهمد و بتواند به زبان روز و عقل منطقی و روانشناسی پاسخ بدهد
