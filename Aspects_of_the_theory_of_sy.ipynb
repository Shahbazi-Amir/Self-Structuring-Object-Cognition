# 📘 **Aspects of the Theory of Syntax** – 1965

---

### 🎯 **هدف کتاب**

* این کتاب ادامه و تکامل ایده‌های *Syntactic Structures* است.
* چامسکی در این اثر چارچوب نظری زبان‌شناسی خودش را **فراتر از نحو** توسعه داد.
* تمرکز اصلی: **چطور کودکان با داده‌های محدود، یک زبان کامل را یاد می‌گیرند؟**

---

### 🧠 **مفاهیم کلیدی کتاب**

#### 1. تمایز **Competence vs Performance**

* **Competence (شایستگی زبانی):** دانش درونی و ناخودآگاه فرد از قواعد زبان.
* **Performance (اجرا):** استفاده واقعی از زبان در گفتار/نوشتار.
* چامسکی می‌گوید: موضوع اصلی زبان‌شناسی باید Competence باشد، نه Performance (چون اجرا می‌تواند خطا، تپق و فراموشی داشته باشد).

---

#### 2. **گرامر جهانی (Universal Grammar – UG)**

* UG مجموعه‌ای از **اصول و قواعد مشترک** است که در همه زبان‌ها وجود دارد.
* این اصول به صورت **فطری** در ذهن انسان وجود دارند.
* تفاوت زبان‌ها ناشی از **پارامترهای قابل تنظیم** است (مثلاً ترتیب فعل و فاعل).

---

#### 3. **Poverty of the Stimulus (فقر داده‌های ورودی)**

* کودکان داده‌های زبانی ناقص و پر از خطا می‌شنوند.
* با وجود این، آن‌ها خیلی سریع و دقیق زبان را یاد می‌گیرند.
* نتیجه: باید یک **دستگاه اکتساب زبان (LAD)** ذاتی وجود داشته باشد که شکاف بین داده‌ها و قواعد را پر کند.

---

#### 4. **Deep Structure vs Surface Structure**

* **Deep Structure (ساختار ژرف):** بازنمایی انتزاعی و معنایی جمله پیش از تبدیل.
* **Surface Structure (ساختار سطحی):** شکل نهایی جمله بعد از اعمال قواعد تبدیلی.

مثال:

* Deep Structure: «Ali is eating something.»
* Surface Structure:

  * جمله خبری: "Ali is eating an apple."
  * جمله پرسشی: "What is Ali eating?"

---

#### 5. **Transformational Rules (قواعد تبدیلی)**

* قواعدی که Deep Structure را به Surface Structure تبدیل می‌کنند.
* این قواعد نشان می‌دهند چرا دو جمله می‌توانند معنا یا شکل مشابه داشته باشند ولی ساختار نحوی متفاوت.

---

#### 6. نقد رفتارگرایی و تقویت **ذاتی‌گرایی**

* چامسکی نشان می‌دهد یادگیری زبان فقط با تقلید و تجربه‌گرایی ممکن نیست.
* زبان **بخشی از زیست‌شناسی انسان** است، نه صرفاً یک مهارت فرهنگی.

---

### 📌 نتیجه‌گیری کتاب

* زبان‌شناسی باید بر **ساختار ذهنی و قواعد فطری زبان** متمرکز شود.
* کودکان یک **LAD** دارند که با ورودی محدود، دستور زبان کامل را می‌سازد.
* زبان یک پدیده‌ی شناختی–زیستی است، نه صرفاً اجتماعی یا رفتاری.

---

### **پیش‌گفتار**

ایدهٔ اینکه زبان بر پایهٔ یک نظام قواعد استوار است که تفسیر بی‌نهایت جملهٔ آن را تعیین می‌کند، به هیچ وجه ایده‌ای نو نیست. بیش از یک قرن پیش، ویلهلم فون هومبولت این دیدگاه را با وضوحی قابل‌قبول در مقدمهٔ مشهور – ولی کمتر مطالعه‌شده – خود بر زبان‌شناسی عمومی (Humboldt, 1836) بیان کرد.

دیدگاه او مبنی بر اینکه زبان "از امکانات محدود، استفاده‌ای نامحدود می‌کند" و اینکه دستور زبان باید فرایندهایی را توصیف کند که این امکان را فراهم می‌آورند، در واقع ریشه در نگرانی دیرینه‌ای دارد در فلسفهٔ عقل‌گرای زبان و ذهن، نسبت به این جنبهٔ «خلاقانه»‌ی استفاده از زبان (برای بحث بیشتر، بنگرید به Chomsky, 1964 و آثار آینده).

حتی می‌توان دستور زبان پانینی را نیز، به‌نوعی، به‌عنوان بخشی از یک «دستور زایشی» تلقی کرد، به معنای امروزی این اصطلاح.

با این حال، در زبان‌شناسی مدرن، تنها در سال‌های اخیر است که تلاش‌هایی نسبتاً قابل توجه برای ساخت دستورهای زایشی صریح برای زبان‌های خاص و بررسی پیامدهای آن‌ها صورت گرفته است. بنابراین، بحث‌ها و مناظرات گسترده پیرامون شکل مناسب نظریهٔ دستور زایشی و توصیف درست زبان‌هایی که بیشتر مورد مطالعه قرار گرفته‌اند، نباید چندان تعجب‌برانگیز باشد.

ماهیت موقتی هرگونه نتیجه‌ای که اکنون بتوان در مورد نظریهٔ زبان‌شناسی یا حتی دستور زبان انگلیسی ارائه کرد، باید برای هر کسی که در این حوزه کار می‌کند، آشکار باشد (برای این کافی‌ست به گسترهٔ وسیع پدیده‌های زبانی بنگریم که تا کنون از صورت‌بندی معنادار در هر قالبی طفره رفته‌اند). با این حال، به نظر می‌رسد برخی نتایج نسبتاً محکم در حال ظهور هستند و به‌طور فزاینده‌ای مورد تأیید قرار می‌گیرند.

به‌ویژه، نقش مرکزی دگرگونی‌های دستوری (grammatical transformations) در هر دستور زایشی که به لحاظ تجربی معتبر باشد، به‌نظر من کاملاً تثبیت شده است، اگرچه هنوز پرسش‌های فراوانی دربارهٔ شکل مناسب نظریهٔ دستور دگرگشتی (transformational grammar) باقی مانده است.

این رساله، مطالعه‌ای اکتشافی دربارهٔ مسائل گوناگونی است که در روند کار روی دستور دگرگشتی پدید آمده‌اند، که این نظریه در سراسر متن به‌عنوان چارچوب کلی بحث مفروض گرفته شده است. پرسش اصلی این است که این نظریه چگونه باید صورت‌بندی شود. بنابراین، این مطالعه به مسائلی می‌پردازد که در مرزهای پژوهش در زمینهٔ دستور دگرگشتی قرار دارند. برای برخی از آن‌ها، پاسخ‌های مشخصی پیشنهاد خواهد شد، اما در اغلب موارد، صرفاً طرح مسئله و بررسی رویکردهای احتمالی انجام خواهد شد، بدون دستیابی به نتیجه‌ای نهایی.

در فصل سوم، من به‌اختصار آنچه را که در پرتو این مباحث، نویدبخش‌ترین جهت‌گیری برای نظریهٔ دستور زایشی می‌دانم، ترسیم خواهم کرد. اما باید تأکید کنم که این صرفاً پیشنهادی بسیار مقدماتی است.

ساختار رساله به‌شرح زیر است:

* فصل اول، پیش‌فرض‌های زمینه‌ای را ترسیم می‌کند. این فصل حاوی نکات جدید چندانی نیست و صرفاً هدف آن خلاصه‌سازی و روشن‌سازی نکات ضروری (و گاهی سوء‌برداشت‌شده) است.

* فصل‌های دوم و سوم به نقص‌های موجود در نسخه‌های اولیهٔ نظریهٔ دستور دگرگشتی می‌پردازند. موقعیت مورد بحث، همان دیدگاه چامسکی (1957)، لیس (1960) و دیگران است. این دیدگاه، مؤلفهٔ نحوی دستور دگرگشتی را متشکل از یک دستور ساخت عبارت به‌عنوان پایه، و مجموعه‌ای از دگرگونی‌ها می‌داند که ساختارهای تولیدشده توسط پایه را به جملات واقعی نگاشت می‌کنند. این دیدگاه در ابتدای فصل سوم به‌اختصار بازگو می‌شود.

* فصل دوم به پایهٔ مؤلفهٔ نحوی و دشواری‌های ناشی از فرض اینکه این پایه صرفاً یک دستور ساخت عبارتی است، می‌پردازد.

* فصل سوم، تجدید نظری در مؤلفهٔ دگرگشتی و رابطهٔ آن با ساختارهای پایه پیشنهاد می‌دهد. مفهوم «دگرگونی دستوری» به‌خودی‌خود بدون تغییر – البته با برخی ساده‌سازی‌ها – حفظ شده است.

* در فصل چهارم، مسائل باقی‌ماندهٔ متعددی مطرح و به‌صورت مختصر و بدون نتیجه‌گیری قطعی بررسی می‌شود.

در پایان، لازم می‌دانم با قدردانی، از دوستان و همکاران زیادی که زحمت خواندن نسخه‌های اولیهٔ این متن را کشیدند و نظرات سودمند خود را در اختیار من گذاشتند، سپاسگزاری کنم.

## **مقدمه‌های روش‌شناختی**

### **§۱. دستورهای زایشی به‌عنوان نظریه‌هایی دربارهٔ توانش زبانی**

این مطالعه به موضوعات متنوعی در نظریهٔ نحوی و نحو زبان انگلیسی می‌پردازد؛ برخی را با جزئیات بیشتر، برخی دیگر را به‌صورت سطحی، و هیچ‌کدام را به‌طور جامع بررسی نمی‌کند. تمرکز این تحقیق بر مؤلفهٔ نحوی یک دستور زایشی است؛ یعنی مجموعه قوانینی که رشته‌های درست و خوش‌ساخت از واحدهای نحوی کمینه (به اصطلاح: «فرماتورها» یا **formatives**) را مشخص می‌کنند و اطلاعات ساختاری گوناگونی را هم به این رشته‌های خوش‌ساخت و هم به رشته‌هایی که از لحاظ نحوی دچار انحراف هستند، نسبت می‌دهند.

چارچوب کلی این پژوهش در منابع متعددی معرفی شده و آشنایی نسبی با مطالعات نظری و توصیفی موجود در کتاب‌نامه، مفروض گرفته می‌شود. در این فصل، من به‌اختصار برخی از مفروضات پس‌زمینه‌ای اصلی را مرور خواهم کرد، بی‌آنکه تلاشی جدی برای اثبات آن‌ها انجام دهم؛ صرفاً با هدف بیان روشن و واضح آن‌ها.

نظریهٔ زبان‌شناسی، در درجهٔ نخست، به گوینده-شنوندهٔ آرمانی می‌پردازد؛ فردی در یک جامعهٔ زبانی کاملاً همگن که زبان خود را به‌صورت کامل می‌داند و تحت تأثیر شرایطی که از نظر دستوری بی‌ربط هستند – مانند محدودیت‌های حافظه، حواس‌پرتی، تغییرات توجه یا علاقه، و خطاهای تصادفی یا الگودار در به‌کارگیری دانش زبانی‌اش در اجراهای واقعی – قرار ندارد.

به نظر من، این دیدگاه، موضع بنیان‌گذاران زبان‌شناسی عمومی مدرن بوده و تا کنون دلیل قانع‌کننده‌ای برای تغییر آن ارائه نشده است. برای مطالعهٔ عملکرد زبانی واقعی، باید به تعامل عوامل مختلف توجه کنیم که توانش زبانی گوینده-شنونده تنها یکی از آن‌هاست. از این نظر، مطالعهٔ زبان تفاوتی با بررسی تجربی پدیده‌های پیچیدهٔ دیگر ندارد.

ما در اینجا تمایزی بنیادی بین **توانش** (دانش گوینده-شنونده از زبان خود) و **اجرا** (کاربرد واقعی زبان در موقعیت‌های عینی) قائل می‌شویم. تنها تحت شرایط آرمانیِ بند بالا، اجرای زبانی بازتاب مستقیمی از توانش زبانی خواهد بود. اما در واقعیت، آشکار است که چنین بازتاب مستقیمی نمی‌تواند وجود داشته باشد. یک نمونهٔ واقعی از گفتار طبیعی، پر است از شروع‌های اشتباه، انحراف از قواعد، تغییر مسیر در میانهٔ بیان و مواردی از این دست.

مسئله برای زبان‌شناس – همان‌طور که برای کودک زبان‌آموز نیز هست – این است که از دل داده‌های عملکرد زبانی، نظام زیرین قواعدی را کشف کند که گوینده-شنونده آن‌ها را آموخته و در اجرا به‌کار می‌گیرد. از این‌رو، **در معنای فنی، نظریهٔ زبان‌شناسی ذهن‌گرایانه (mentalist) است**، چرا که هدفش کشف واقعیتی ذهنی در پس رفتار زبانیِ قابل مشاهده است.

کاربرد واقعی زبان، یا آمادگی‌های فرضی برای پاسخ‌دادن، یا عادات زبانی، شاید بتوانند **شواهدی دربارهٔ ماهیت این واقعیت ذهنی** فراهم کنند، اما قطعاً نمی‌توانند خودِ موضوع اصلی علم زبان‌شناسی را تشکیل دهند – دست‌کم اگر این علم بخواهد جدی گرفته شود.

این تمایزی که در اینجا شرح داده شده، با تمایز **لانگ (langue)** و **پارول (parole)** نزد سوسور مرتبط است؛ ولی ما ناچاریم مفهوم *langue* را نزد سوسور – به‌عنوان صرفاً فهرستی نظام‌مند از اقلام زبانی – کنار بگذاریم و به برداشت **هومبولتی** از توانش به‌عنوان نظامی از **فرایندهای زایشی** بازگردیم (برای بحث بیشتر، بنگرید به Chomsky, 1964).

دستور زبان یک زبان، تلاشی است برای توصیف توانش ذاتی گوینده-شنوندهٔ آرمانی. اگر این دستور، علاوه بر این، کاملاً صریح باشد – به‌عبارتی، اگر فهم آن متکی بر هوش خواننده نباشد و تحلیل صریحی از سهم مشارکت ذهنی او ارائه دهد – می‌توان آن را **دستور زایشی (generative grammar)** نامید (هرچند این اصطلاح کمی تکراری به نظر برسد).

یک دستور زبان کاملاً کارآمد، باید برای هر جمله از مجموعهٔ بی‌نهایت جملات ممکن، **توصیف ساختاری‌ای ارائه کند که نشان دهد آن جمله توسط گوینده-شنوندهٔ آرمانی چگونه فهمیده می‌شود**. این همان مسئلهٔ سنتی زبان‌شناسی توصیفی است؛ و دستورهای سنتی نیز اطلاعات فراوانی دربارهٔ توصیف‌های ساختاری جملات در اختیار می‌گذارند.

با وجود ارزشمندی آشکار آن‌ها، اما دستورهای سنتی در این زمینه **نقص‌هایی دارند که در ادامه بررسی خواهد شد**…

---

اگر خواستی ادامهٔ این بخش یا کل فصل رو هم ترجمه کنم، فقط بگو. همچنین می‌تونم همهٔ ترجمه‌ها رو در یک فایل مارک‌داون یا PDF برای استفاده در پروژه‌هات آماده کنم.
با کمال میل، امیر. در ادامه ترجمهٔ کامل این بخش از فصل اول (ادامهٔ §۱ از *Methodological Preliminaries*) آورده شده است:

---

...دستورهای سنتی، بسیاری از نظم‌ها و قواعد اساسی زبان مورد بحث خود را بیان نمی‌کنند. این واقعیت به‌ویژه در سطح نحو (Syntax) آشکار است، جایی که هیچ دستور زبان سنتی یا ساخت‌گرایانه‌ای از طبقه‌بندی نمونه‌های خاص فراتر نمی‌رود تا به مرحلهٔ صورت‌بندی قواعد زایشی در مقیاسی معنادار برسد.

تحلیل دقیق بهترین دستورهای موجود خیلی سریع نشان می‌دهد که این یک **نقص بنیادی** است، نه صرفاً یک ایراد در جزئیات تجربی یا دقت منطقی. با این حال، به نظر می‌رسد تلاشی که برای کاوش این قلمرو تا حد زیادی ناشناخته صورت می‌گیرد، می‌تواند سودمندانه با مطالعهٔ نوع اطلاعات ساختاری ارائه‌شده در دستورهای سنتی و فرایندهای زبانی‌ای که – هرچند غیررسمی – در این دستورها منعکس شده‌اند، آغاز شود.

باید **محدودیت‌های دستورهای سنتی و ساخت‌گرایانه را به‌خوبی درک کرد**. گرچه این دستورها ممکن است فهرست‌های کامل و دقیقی از استثناها و بی‌قاعدگی‌ها را دربرداشته باشند، اما تنها نمونه‌ها و اشاراتی کلی نسبت به فرایندهای نحوی منظم و مولد ارائه می‌دهند.

نظریهٔ سنتی زبان‌شناسی از این موضوع بی‌اطلاع نبود. مثلاً، **جیمز بیتی** (James Beattie) در سال ۱۷۸۸ می‌نویسد:

> «زبان‌ها از این نظر شبیه انسان‌ها هستند که هر یک ویژگی‌هایی دارند که آن‌ها را از دیگر زبان‌ها متمایز می‌کند، اما همگی صفاتی مشترک نیز دارند. ویژگی‌های منحصر‌به‌فرد زبان‌های مختلف، در دستورها و واژه‌نامه‌های آن زبان‌ها توضیح داده می‌شوند.
> اما آن دسته از ویژگی‌هایی که در همهٔ زبان‌ها مشترک‌اند، یا برای هر زبان ضرورت دارند، در دانشی بررسی می‌شوند که برخی آن را "دستور زبان عام" یا "دستور زبان فلسفی" می‌نامند.»

پیش‌تر از او، **دو مارسِه (Du Marsais)** در سال ۱۷۲۹، دستور زبان عام و خاص را این‌گونه تعریف می‌کند (به نقل از Sahlin, 1928، صفحات ۲۹–۳۰):

> «در دستور زبان، مشاهداتی وجود دارند که در مورد همهٔ زبان‌ها صدق می‌کنند؛ این مشاهدات همان چیزی هستند که به آن *دستور زبان عام* گفته می‌شود: مثل مشاهداتی دربارهٔ آواهای گفتاری، حروفی که نماد این آواها هستند، ماهیت واژگان، و روش‌های گوناگونی که واژگان باید با آن‌ها چیده یا صرف شوند تا معنا بسازند.
> علاوه بر این مشاهدات عام، مشاهداتی هم وجود دارند که خاص یک زبان معین هستند؛ و این همان چیزی است که *دستور زبان‌های خاص* را برای هر زبان تشکیل می‌دهد.»

در درون نظریهٔ سنتی زبان‌شناسی، این نکته نیز به‌وضوح درک شده بود که یکی از ویژگی‌های مشترک همهٔ زبان‌ها، **جنبهٔ "خلاقانه"‌ی آن‌ها**ست. به‌بیان دیگر، ویژگی ذاتی زبان آن است که ابزارهایی فراهم می‌کند برای بیان تعداد بی‌نهایتی از اندیشه‌ها و برای واکنش مناسب در دامنه‌ای بی‌پایان از موقعیت‌های جدید (برای منابع بیشتر، بنگرید به Chomsky, 1964 و آثار آتی او).

از این‌رو، دستور زبان یک زبان خاص باید **با یک "دستور زبان عام" تکمیل شود** که بتواند جنبهٔ خلاقانهٔ زبان و نظم‌های عمیق و بنیادینی را که به‌دلیل جهان‌شمول بودن، در دستور زبان خاص ذکر نشده‌اند، پوشش دهد.

بنابراین، کاملاً منطقی است که یک دستور زبان تنها به استثناها و بی‌قاعدگی‌ها بپردازد. تنها وقتی که با یک دستور زبان عام پشتیبانی شود، دستور زبان یک زبان می‌تواند **توصیف کاملی از توانش زبانی گوینده-شنونده** ارائه دهد.

با این حال، **زبان‌شناسی مدرن به‌طور صریح، ضرورت تکمیل یک "دستور زبان خاص" با یک "دستور زبان عام" را برای رسیدن به کفایت توصیفی به رسمیت نشناخته است.** در واقع، این رشته به‌طور مشخص مطالعهٔ دستور زبان عام را بی‌راهه تلقی کرده است؛ و همان‌طور که پیش‌تر اشاره شد، تلاشی هم برای مواجهه با جنبهٔ خلاقانهٔ کاربرد زبان صورت نداده است. از این‌رو، راه‌حلی برای غلبه بر نقص توصیفی بنیادین دستورهای ساخت‌گرا ارائه نمی‌دهد.

یکی دیگر از دلایل شکست دستورهای سنتی – چه خاص و چه عام – در بیان دقیق فرایندهای منظمِ ساخت و تفسیر جمله‌ها، **باور رایجی بود که "ترتیب واژگان، بازتابی از ترتیب طبیعی افکار است."** بر اساس این دیدگاه، قواعد مربوط به ساخت جمله‌ها واقعاً متعلق به علم دستور زبان نیستند، بلکه به رشته‌ای دیگر مربوط می‌شوند که به "ترتیب افکار" می‌پردازد.

این دیدگاه در اثر مشهوری به نام **Grammaire générale et raisonnée** نیز بازتاب یافته است…

حتماً امیر جان، در ادامه ترجمهٔ کامل پاراگراف پایانی بخش §1 از فصل *Methodological Preliminaries* ارائه شده است:

---

در کتاب *Grammaire générale et raisonnée* اثر **لانسلوت و دیگران (1660)** آمده است که – به‌جز در گفتار مجازی – **ترتیب واژگان تابع "نظم طبیعی" است** که با «بیان طبیعی اندیشه‌های ما» سازگار است. از این‌رو، تنها به تعداد اندکی از قواعد دستوری مانند حذف (ellipsis)، وارونگی (inversion) و امثال آن نیاز داریم، یعنی قواعدی که استفادهٔ مجازی از زبان را تعیین می‌کنند.

این دیدگاه به اشکال و گونه‌های مختلف در سنت فکری زبان‌شناسی حضور دارد. به عنوان مثال دیگر، **دیدرو (Diderot)** در مقاله‌ای جالب – که عمدتاً به این پرسش می‌پردازد که چگونه نظم هم‌زمان و پی‌درپی اندیشه‌ها در ترتیب واژگان بازتاب می‌یابد – به این نتیجه می‌رسد که زبان فرانسوی، بیش از هر زبان دیگری، ترتیب واژگانش با ترتیب طبیعی افکار و اندیشه‌ها منطبق است (Diderot, 1751). او می‌نویسد:

> «صرف‌نظر از ترتیب واژگان در زبان‌های باستانی یا مدرن، ذهن نویسنده همیشه از نظم آموزشی دستور زبان فرانسوی پیروی کرده است» (ص. ۳۹۰)،
> و «ما در زبان فرانسوی چیزها را همان‌گونه بیان می‌کنیم که ذهن مجبور است آن‌ها را در هر زبانی در نظر بگیرد» (ص. ۳۷۱).

دیدرو با انسجامی ستودنی نتیجه می‌گیرد که:

> «زبان ساده و پیاده‌روار ما نسبت به سایر زبان‌ها برتری کاربردی دارد نه زیبایی‌شناختی» (ص. ۳۷۲).
> بنابراین، زبان فرانسوی برای علوم مناسب است، در حالی که یونانی، لاتین، ایتالیایی و انگلیسی «برای ادبیات مناسب‌ترند.»

او حتی می‌افزاید:

> «عقل سلیم زبان فرانسوی را انتخاب می‌کند؛ اما... تخیل و احساسات، زبان‌های باستانی و زبان‌های همسایگان‌مان را ترجیح خواهند داد... باید در جامعه و در مدارس فلسفه به زبان فرانسوی صحبت کرد؛ و در خطابه‌ها و تئاترها به یونانی، لاتین یا انگلیسی... زبان ما، زبان حقیقت خواهد بود، اگر روزی حقیقت دوباره به زمین بازگردد؛ و زبان یونانی، لاتین و دیگر زبان‌ها، زبان‌های اسطوره و دروغ خواهند بود.
> زبان فرانسوی برای آموزش، روشنگری و اقناع ساخته شده است؛ یونانی، لاتین، ایتالیایی و انگلیسی برای ترغیب، تحریک و فریب: با مردم به یونانی، لاتین یا ایتالیایی سخن بگویید؛ اما با خردمند، فرانسوی صحبت کنید.» (ص. ۳۷۱–۳۷۲)

به‌هرحال، تا جایی که ترتیب واژگان به عواملی مستقل از زبان وابسته باشد، نیازی به توصیف آن در دستور زبان خاص یا عام نیست. در نتیجه، **دلیل اصولی برای حذف صریح قواعد نحوی از دستور زبان وجود دارد.**

شایان ذکر است که این نگاه ساده‌انگارانه به ساختار زبان، **تا دوران معاصر نیز در اشکال مختلفی تداوم یافته است**؛ مثلاً در تصویر **سوسور** از دنباله‌ای از نمادهای زبانی که با زنجیره‌ای بی‌شکل از مفاهیم متناظرند، یا در دیدگاه رایجی که استفاده از زبان را صرفاً استفاده از واژگان و عبارات می‌داند (مثلاً در آثار **رایل، ۱۹۵۳**).

اما **علت بنیادی ناکارآمدی دستور زبان‌های سنتی، فنی‌تر از این‌هاست.**
با اینکه به‌خوبی درک شده بود که فرایندهای زبانی از جهاتی «خلاقانه» هستند، **ابزارهای فنی لازم برای صورت‌بندی نظامی بازگشتی (recursive)** هنوز در اختیار نبودند…
